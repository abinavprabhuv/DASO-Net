```python

# DASO-Net — CONTINUATION DATASET DIAGNOSTIC v1
# Read-only diagnostic. No extraction, training, or model inference.

from pathlib import Path
import hashlib
import zipfile
import os

ROOT = Path(
    "/kaggle/input/datasets/abinavprabhuv/"
    "dasonet-continuation-checkpoint-v1"
)

EXPECTED_SHA256 = (
    "39bd2696766eff9b3597493cd17eca3a73a967c5223bcf085f51773430a4762b"
)
EXPECTED_SIZE = 39_887_900

def sha256_file(path):
    digest = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(8 * 1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()

print("=" * 76)
print("DASO-Net CONTINUATION DATASET DIAGNOSTIC")
print("=" * 76)
print("Dataset root:", ROOT)
print("Root exists:", ROOT.exists())

if not ROOT.exists():
    print("\nSTOP: dataset root is missing.")
    print("Attached dataset roots:")
    for p in sorted(Path("/kaggle/input").glob("**/")):
        if p.parent == Path("/kaggle/input/datasets/abinavprabhuv"):
            print(" ", p)
else:
    files = sorted(p for p in ROOT.rglob("*") if p.is_file())

    print("\nTotal files found:", len(files))
    print("\nFULL FILE INVENTORY")
    print("-" * 76)

    if not files:
        print("No files found inside the attached dataset.")

    for path in files:
        rel = path.relative_to(ROOT)
        size = path.stat().st_size
        print(f"{size:>14,} bytes | {rel}")

    print("\nARCHIVE AND CHECKSUM INSPECTION")
    print("-" * 76)

    archive_count = 0
    exact_hash_matches = []
    expected_size_matches = []

    for path in files:
        is_zip = path.suffix.lower() == ".zip"

        # Also detect ZIP data whose filename does not end in .zip.
        if not is_zip:
            try:
                is_zip = zipfile.is_zipfile(path)
            except OSError:
                is_zip = False

        is_checksum = (
            path.name.lower().endswith((".sha256", ".sha256.txt", ".txt"))
            and ("sha256" in path.name.lower() or "checksum" in path.name.lower())
        )

        if is_zip:
            archive_count += 1
            print(f"\nZIP candidate: {path}")
            print(f"  Size:   {path.stat().st_size:,} bytes")
            print(f"  SHA256: {sha256_file(path)}")

            if path.stat().st_size == EXPECTED_SIZE:
                expected_size_matches.append(path)

            if sha256_file(path) == EXPECTED_SHA256:
                exact_hash_matches.append(path)
                print("  RESULT: EXACT FROZEN SHA-256 MATCH")
            else:
                print("  RESULT: does not match frozen ZIP SHA-256")

        elif is_checksum:
            print(f"\nChecksum/readme file: {path}")
            try:
                print(path.read_text(encoding="utf-8", errors="replace")[:4000])
            except OSError as exc:
                print("  Could not read:", exc)

    print("\n" + "=" * 76)
    print("DIAGNOSTIC SUMMARY")
    print("=" * 76)
    print("Files found:", len(files))
    print("ZIP archives found:", archive_count)
    print("Files with expected ZIP size:", len(expected_size_matches))
    print("Exact frozen ZIP SHA-256 matches:", len(exact_hash_matches))

    if exact_hash_matches:
        print("\nPASS: The original continuation ZIP is present.")
        print("Verified archive:", exact_hash_matches[0])
        print("Do not extract it in this diagnostic cell; we will use")
        print("a corrected complete bootstrap cell next.")
    elif archive_count:
        print("\nSTOP: ZIP files exist, but none matches the frozen archive.")
        print("Review the inventory and checksum output before proceeding.")
    else:
        print("\nSTOP: No ZIP archive was found.")
        print("The Kaggle dataset may contain extracted files, a renamed")
        print("non-ZIP package, or an incomplete upload.")

    print("\nTraining performed: NO")
    print("Model inference performed: NO")
    print("Frozen input files modified: NO")
```

```python

# DASO-Net — EXTRACTED CONTINUATION VERIFICATION + TIOMAN SOURCE AUDIT v2
# Read-only inputs. No extraction, training, inference, or metric computation.
# Creates a new audit directory and preserves all earlier outputs.

from pathlib import Path
import csv
import hashlib
import json
import os
import re
from collections import Counter

# ---------------------------------------------------------------------
# 1. Known mounted dataset roots
# ---------------------------------------------------------------------
PROJECT_ROOT = Path(
    "/kaggle/input/datasets/abinavprabhuv/dasonet-project"
)
CONTINUATION_ROOT = Path(
    "/kaggle/input/datasets/abinavprabhuv/"
    "dasonet-continuation-checkpoint-v1"
)
TIOMAN_ROOT = Path(
    "/kaggle/input/datasets/abinavprabhuv/dasonet-tioman-source-v1"
)
WORK_ROOT = Path("/kaggle/working")

EXPECTED_FREEZE_SHA256 = (
    "228935b43de2e47a1022110eeb2cd5e012a3d0b705a1f4e519af2e0904579902"
)
EXPECTED_CONTINUATION_ZIP_SHA256 = (
    "39bd2696766eff9b3597493cd17eca3a73a967c5223bcf085f51773430a4762b"
)
EXPECTED_TIOMAN_AUDIT_SHA256 = (
    "cf99b32c23659182d1de879559e62655072dc68018181f0e2c623c62cc8d09f7"
)

EXPECTED_CHECKPOINTS = {
    "seed01_s1782656963": (
        "e4c4b79ad4b8c6e18ba2b1e3d6cbe796c9cc301bd6fcc7c596fc290ad3a7e81d"
    ),
    "seed02_s901395089": (
        "19cdb834a46240916b2c8023e8c3afe5738d9c7cd5d8f818c62ad1e289df1ad2"
    ),
    "seed03_s252916394": (
        "be7b7941f2f62179e15074aa973bf6da0eb9018ec9133486512bd1f87153f447"
    ),
}

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".tif", ".tiff", ".bmp", ".webp"
}


# ---------------------------------------------------------------------
# 2. Helpers
# ---------------------------------------------------------------------
def sha256_file(path):
    digest = hashlib.sha256()
    with open(path, "rb") as stream:
        for chunk in iter(lambda: stream.read(8 * 1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()


def parse_sha256_text(path):
    if not path or not path.is_file():
        return None
    content = path.read_text(encoding="utf-8", errors="replace")
    match = re.search(r"\b([0-9a-fA-F]{64})\b", content)
    return match.group(1).lower() if match else None


def sidecar_for(path):
    candidates = [
        path.with_suffix(".sha256.txt"),
        Path(str(path) + ".sha256.txt"),
    ]
    return next((p for p in candidates if p.is_file()), None)


def new_output_dir(base):
    candidate = base
    number = 2
    while candidate.exists():
        candidate = base.with_name(f"{base.name}_run{number:02d}")
        number += 1
    candidate.mkdir(parents=True, exist_ok=False)
    return candidate


def norm_key(value):
    return re.sub(r"[^a-z0-9]+", "", str(value).lower())


def is_sha256(value):
    return isinstance(value, str) and bool(
        re.fullmatch(r"[0-9a-fA-F]{64}", value.strip())
    )


def unique_files(root, filename=None, suffix=None):
    if not root.exists():
        return []
    results = []
    for path in root.rglob("*"):
        if not path.is_file():
            continue
        if filename is not None and path.name != filename:
            continue
        if suffix is not None and path.suffix.lower() != suffix.lower():
            continue
        results.append(path)
    return sorted(set(results))


# ---------------------------------------------------------------------
# 3. Check dataset mounts
# ---------------------------------------------------------------------
print("=" * 78)
print("DASO-Net — EXTRACTED CONTINUATION VERIFICATION + TIOMAN AUDIT v2")
print("=" * 78)

for label, root in [
    ("Frozen project", PROJECT_ROOT),
    ("Continuation dataset", CONTINUATION_ROOT),
    ("Tioman source dataset", TIOMAN_ROOT),
]:
    print(f"{label}: {root}")
    print(f"  Exists: {root.is_dir()}")

if not PROJECT_ROOT.is_dir():
    raise RuntimeError("STOP: frozen project dataset is not mounted.")
if not CONTINUATION_ROOT.is_dir():
    raise RuntimeError("STOP: continuation dataset is not mounted.")
if not TIOMAN_ROOT.is_dir():
    raise RuntimeError("STOP: Tioman source dataset is not mounted.")


# ---------------------------------------------------------------------
# 4. Identify the packaging layout; do not require the ZIP to exist
# ---------------------------------------------------------------------
exact_zip_files = unique_files(
    CONTINUATION_ROOT,
    filename="DASO_NET_CONTINUATION_CHECKPOINT_v1.zip",
)
zip_sidecars = unique_files(
    CONTINUATION_ROOT,
    filename="DASO_NET_CONTINUATION_CHECKPOINT_v1.zip.sha256.txt",
)

print("\nCONTINUATION PACKAGING")
if exact_zip_files:
    print("Original ZIP file found:")
    for path in exact_zip_files:
        print(" ", path, "| SHA-256:", sha256_file(path))
    zip_present = True
else:
    print("Original continuation ZIP: NOT PRESENT")
    print("Extracted-file verification will be used instead.")
    zip_present = False

zip_sidecar_hashes = [
    parse_sha256_text(path) for path in zip_sidecars
]
zip_sidecar_matches = EXPECTED_CONTINUATION_ZIP_SHA256 in zip_sidecar_hashes

print("ZIP checksum sidecar present:", bool(zip_sidecars))
print("ZIP sidecar contains the recorded expected hash:", zip_sidecar_matches)
print(
    "Note: a checksum sidecar is not proof of the missing ZIP's bytes. "
    "The extracted contents will be checked independently."
)


# ---------------------------------------------------------------------
# 5. Verify the frozen continuation record by its exact SHA-256
# ---------------------------------------------------------------------
freeze_candidates = unique_files(
    CONTINUATION_ROOT,
    filename="DASONET_CONTINUATION_CHECKPOINT_FREEZE_v1.json",
)

freeze_matches = [
    path for path in freeze_candidates
    if sha256_file(path) == EXPECTED_FREEZE_SHA256
]

print("\nFROZEN CONTINUATION RECORD")
for path in freeze_candidates:
    print(f"  {path}")
    print(f"    SHA-256: {sha256_file(path)}")

if not freeze_matches:
    raise RuntimeError(
        "STOP: the frozen continuation JSON was not found with its expected "
        "SHA-256. Do not train or evaluate until this is resolved."
    )

freeze_path = freeze_matches[0]
print("PASS: exact frozen continuation record verified.")


# Verify the file-manifest JSON against its own SHA sidecar when available.
file_manifest_candidates = unique_files(
    CONTINUATION_ROOT,
    filename="DASONET_CONTINUATION_CHECKPOINT_FILE_MANIFEST_v1.json",
)
file_manifest_checks = []

for path in file_manifest_candidates:
    sidecar = sidecar_for(path)
    actual_sha = sha256_file(path)
    expected_sidecar_sha = parse_sha256_text(sidecar) if sidecar else None
    passed = bool(expected_sidecar_sha and actual_sha == expected_sidecar_sha)
    file_manifest_checks.append({
        "path": str(path),
        "sidecar_path": str(sidecar) if sidecar else None,
        "actual_sha256": actual_sha,
        "sidecar_sha256": expected_sidecar_sha,
        "passed": passed,
    })
    print("\nContinuation file manifest:", path)
    print("  Sidecar found:", bool(sidecar))
    print("  JSON SHA-256:", actual_sha)
    print("  Sidecar SHA-256:", expected_sidecar_sha)
    print("  Sidecar verification:", "PASS" if passed else "NOT VERIFIED")

if file_manifest_candidates and not any(
    item["passed"] for item in file_manifest_checks
):
    raise RuntimeError(
        "STOP: the continuation file-manifest JSON did not match its "
        "SHA-256 sidecar. Preserve the output and investigate."
    )


# ---------------------------------------------------------------------
# 6. Verify all three frozen checkpoint files by their content hashes
# ---------------------------------------------------------------------
checkpoint_candidates = unique_files(CONTINUATION_ROOT, suffix=".pt")
checkpoint_hash_index = {}

print("\nCHECKPOINT HASH AUDIT")
print("Checkpoint files found:", len(checkpoint_candidates))

for path in checkpoint_candidates:
    digest = sha256_file(path)
    checkpoint_hash_index.setdefault(digest, []).append(str(path))
    print(f"  {path.name} | {path.stat().st_size:,} bytes | {digest}")

verified_checkpoints = {}
checkpoint_failures = []

for seed_name, expected_sha in EXPECTED_CHECKPOINTS.items():
    matches = checkpoint_hash_index.get(expected_sha, [])
    if matches:
        verified_checkpoints[seed_name] = matches
        print(f"PASS {seed_name}")
        for match in matches:
            print(" ", match)
    else:
        checkpoint_failures.append(seed_name)
        print(f"FAIL {seed_name}: expected SHA-256 was not found.")

if checkpoint_failures:
    raise RuntimeError(
        "STOP: one or more frozen checkpoints failed exact hash verification: "
        + ", ".join(checkpoint_failures)
    )

print("CHECKPOINT VERIFICATION: PASS — all three frozen checkpoints match.")


# ---------------------------------------------------------------------
# 7. Collect frozen Tioman expected hashes from preserved records
# ---------------------------------------------------------------------
# The continuation package includes previous Tioman audit records and the
# source-specific public-real manifest. Search those records; do not infer
# expected hashes from the newly uploaded images themselves.

expected_hash_candidates = []


def add_candidate(source, records):
    clean_records = []
    for record in records:
        digest = str(record.get("sha256", "")).strip().lower()
        if is_sha256(digest):
            clean_records.append({
                "filename": str(record.get("filename", "") or ""),
                "sha256": digest,
            })

    unique_hashes = {item["sha256"] for item in clean_records}
    if unique_hashes:
        expected_hash_candidates.append({
            "source": str(source),
            "records": clean_records,
            "hashes": unique_hashes,
            "unique_hash_count": len(unique_hashes),
        })


def json_sha_records(obj):
    """Extract hash/path pairs from nested JSON objects."""
    records = []

    if isinstance(obj, dict):
        sha_values = []
        image_names = []

        for key, value in obj.items():
            normalized = norm_key(key)

            if (
                isinstance(value, str)
                and "sha256" in normalized
                and is_sha256(value.strip())
            ):
                sha_values.append(value.strip().lower())

            if isinstance(value, str):
                candidate_name = value.strip()
                if Path(candidate_name).suffix.lower() in IMAGE_EXTENSIONS:
                    image_names.append(candidate_name)

        if sha_values:
            # Keep a record for every SHA even if no filename field is stored.
            if image_names:
                for digest in sha_values:
                    for name in image_names:
                        records.append({
                            "filename": name,
                            "sha256": digest,
                        })
            else:
                for digest in sha_values:
                    records.append({
                        "filename": "",
                        "sha256": digest,
                    })

        for value in obj.values():
            records.extend(json_sha_records(value))

    elif isinstance(obj, list):
        for value in obj:
            records.extend(json_sha_records(value))

    return records


# First: inspect the exact previously frozen Tioman asset-resolution audit.
tioman_audit_files = unique_files(
    CONTINUATION_ROOT,
    filename="TIOMAN_EXACT_ASSET_RESOLUTION_AUDIT_v1.json",
)

for path in tioman_audit_files:
    actual_sha = sha256_file(path)
    print("\nPreserved Tioman asset-resolution audit:", path)
    print("  SHA-256:", actual_sha)

    if actual_sha == EXPECTED_TIOMAN_AUDIT_SHA256:
        print("  Audit-record integrity: PASS")
    else:
        print("  Audit-record integrity: DOES NOT MATCH EXPECTED SHA")

    try:
        obj = json.loads(path.read_text(encoding="utf-8"))
        records = json_sha_records(obj)
        add_candidate(path, records)
    except Exception as exc:
        print("  Could not parse audit JSON:", repr(exc))


# Second: inspect frozen CSV/TSV manifests in the continuation bundle
# and frozen project. Include the known source-specific manifest by name.
manifest_roots = [CONTINUATION_ROOT, PROJECT_ROOT]
seen_manifest_paths = set()

for root in manifest_roots:
    for path in root.rglob("*"):
        if not path.is_file() or path.suffix.lower() not in {".csv", ".tsv"}:
            continue

        is_source_manifest = (
            path.name == "public_real_mp_v1_source_specific_manifest.csv"
        )

        try:
            if path.stat().st_size > 100 * 1024 * 1024:
                continue

            raw = path.read_bytes()
            if not is_source_manifest and b"tioman" not in raw.lower():
                continue

            if str(path) in seen_manifest_paths:
                continue
            seen_manifest_paths.add(str(path))

            text = raw.decode("utf-8-sig", errors="replace")
            delimiter = "\t" if path.suffix.lower() == ".tsv" else ","
            reader = csv.DictReader(text.splitlines(), delimiter=delimiter)
            headers = reader.fieldnames or []

            sha_columns = [
                header for header in headers
                if "sha256" in norm_key(header)
            ]
            if not sha_columns:
                continue

            filename_columns = [
                header for header in headers
                if any(token in norm_key(header) for token in (
                    "filename", "filepath", "relativepath",
                    "imagepath", "imagefile", "sourcefile", "assetpath"
                ))
            ]

            records = []
            for row in reader:
                row_text = " ".join(str(v or "") for v in row.values())
                if "tioman" not in row_text.lower():
                    continue

                digest = None
                for column in sha_columns:
                    value = str(row.get(column, "") or "").strip().lower()
                    value = value.removeprefix("sha256:")
                    if is_sha256(value):
                        digest = value
                        break

                if not digest:
                    continue

                filename = ""
                for column in filename_columns:
                    value = str(row.get(column, "") or "").strip()
                    if value:
                        filename = value
                        break

                if not filename:
                    for value in row.values():
                        value = str(value or "").strip()
                        if Path(value).suffix.lower() in IMAGE_EXTENSIONS:
                            filename = value
                            break

                records.append({
                    "filename": filename,
                    "sha256": digest,
                })

            if records:
                add_candidate(path, records)

        except Exception as exc:
            print("Manifest inspection warning:", path, repr(exc))


print("\nTIOMAN EXPECTED-HASH CANDIDATES")
if not expected_hash_candidates:
    print("No usable candidate hash records were extracted.")
else:
    for item in expected_hash_candidates:
        print(
            f"  {item['source']} | records={len(item['records'])} "
            f"| unique SHA-256={item['unique_hash_count']}"
        )

exact_35_candidates = [
    item for item in expected_hash_candidates
    if item["unique_hash_count"] == 35
]

expected_hashes = set()
expected_records = []
expected_manifest_source = None
expected_manifest_status = "NOT_RESOLVED"

if exact_35_candidates:
    distinct_sets = {
        frozenset(item["hashes"]) for item in exact_35_candidates
    }

    if len(distinct_sets) == 1:
        selected = exact_35_candidates[0]
        expected_hashes = set(selected["hashes"])
        expected_records = selected["records"]
        expected_manifest_source = selected["source"]
        expected_manifest_status = "PASS_35_UNIQUE_FROZEN_HASHES"
        print("\nExpected-hash source selected:", expected_manifest_source)
        print("Expected unique Tioman hashes:", len(expected_hashes))
    else:
        expected_manifest_status = "STOP_CONFLICTING_FROZEN_HASH_SETS"
        print(
            "\nSTOP: multiple frozen sources contain 35 hashes, but the "
            "hash sets disagree. No source has been selected."
        )
else:
    expected_manifest_status = "STOP_35_HASHES_NOT_RESOLVED"
    print(
        "\nSTOP: no preserved source produced exactly 35 unique expected "
        "image hashes. Do not infer from the Tioman images yet."
    )


# ---------------------------------------------------------------------
# 8. Hash actual image files in the mounted Tioman dataset
# ---------------------------------------------------------------------
tioman_image_files = [
    path for path in TIOMAN_ROOT.rglob("*")
    if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS
]

actual_hash_to_paths = {}

print("\nTIOMAN SOURCE INVENTORY")
print("Image files found:", len(tioman_image_files))

for path in tioman_image_files:
    digest = sha256_file(path)
    actual_hash_to_paths.setdefault(digest, []).append(str(path))

print("Unique image SHA-256 values:", len(actual_hash_to_paths))
print("Image extension counts:", dict(Counter(
    path.suffix.lower() for path in tioman_image_files
)))


# ---------------------------------------------------------------------
# 9. Exact 35-image hash comparison
# ---------------------------------------------------------------------
matched_hashes = set()
missing_hashes = set()

if expected_manifest_status == "PASS_35_UNIQUE_FROZEN_HASHES":
    matched_hashes = expected_hashes.intersection(actual_hash_to_paths)
    missing_hashes = expected_hashes.difference(actual_hash_to_paths)

    if len(matched_hashes) == 35 and not missing_hashes:
        tioman_status = "PASS_35_OF_35_EXACT_SHA256_MATCHES"
    else:
        tioman_status = "STOP_TiOMAN_SOURCE_HASH_MISMATCH"
else:
    tioman_status = expected_manifest_status

print("\n" + "=" * 78)
print("TIOMAN EXACT SOURCE VERIFICATION")
print("=" * 78)
print("Status:", tioman_status)
print("Expected unique hashes:", len(expected_hashes))
print("Exact hash matches:", len(matched_hashes))
print("Expected hashes not found:", len(missing_hashes))

if missing_hashes:
    print("\nMissing expected image hashes:")
    for digest in sorted(missing_hashes):
        names = sorted({
            Path(item["filename"]).name
            for item in expected_records
            if item["sha256"] == digest and item["filename"]
        })
        print(f"  {digest} | expected name(s): {names}")

if tioman_status == "PASS_35_OF_35_EXACT_SHA256_MATCHES":
    print("\nPASS: all 35 frozen expected image hashes match exactly.")
else:
    print("\nSTOP: do not run Tioman inference until the audit passes.")


# ---------------------------------------------------------------------
# 10. Save a new audit record without overwriting previous attempts
# ---------------------------------------------------------------------
audit_dir = new_output_dir(
    WORK_ROOT / "dasonet_extracted_continuation_tioman_audit_v2"
)

audit_record = {
    "audit_name": "DASO-Net extracted continuation and Tioman source audit v2",
    "training_performed": False,
    "inference_performed": False,
    "frozen_project_root": str(PROJECT_ROOT),
    "continuation_dataset_root": str(CONTINUATION_ROOT),
    "tioman_source_root": str(TIOMAN_ROOT),
    "continuation_payload_mode": (
        "original_zip_present" if zip_present else "extracted_contents"
    ),
    "original_continuation_zip_present": zip_present,
    "continuation_zip_sidecar_matches_recorded_sha": zip_sidecar_matches,
    "continuation_zip_sha256_actually_verified": bool(zip_present),
    "freeze_json_path": str(freeze_path),
    "freeze_json_sha256": sha256_file(freeze_path),
    "freeze_json_verified": True,
    "continuation_file_manifest_checks": file_manifest_checks,
    "verified_checkpoints": verified_checkpoints,
    "tioman_expected_hash_source": expected_manifest_source,
    "tioman_expected_hash_status": expected_manifest_status,
    "tioman_expected_unique_hash_count": len(expected_hashes),
    "tioman_source_image_file_count": len(tioman_image_files),
    "tioman_source_unique_image_hash_count": len(actual_hash_to_paths),
    "tioman_exact_hash_match_count": len(matched_hashes),
    "tioman_missing_expected_hashes": sorted(missing_hashes),
    "tioman_verification_status": tioman_status,
    "next_step": (
        "Preserve this audit and prepare the frozen Tioman external evaluation."
        if tioman_status == "PASS_35_OF_35_EXACT_SHA256_MATCHES"
        else
        "Resolve the manifest/hash issue before any Tioman inference."
    ),
}

audit_json = audit_dir / "DASONET_EXTRACTED_CONTINUATION_TIOMAN_AUDIT_v2.json"
audit_json.write_text(
    json.dumps(audit_record, indent=2, sort_keys=True),
    encoding="utf-8",
)

print("\n" + "=" * 78)
print("AUDIT RECORD SAVED")
print("=" * 78)
print("Audit directory:", audit_dir)
print("Audit JSON:", audit_json)
print("Audit JSON SHA-256:", sha256_file(audit_json))
print("Training performed: NO")
print("Model inference performed: NO")
print("Frozen input datasets modified: NO")
print("Tioman status:", tioman_status)
```

```python

# ================================================================
# DASO-Net — TIOMAN FTIR DOCUMENT CONTENT AUDIT v1
#
# PURPOSE
#   1. Inventory and SHA-256 hash Tioman JPG, PDF and ISPD files.
#   2. Test exact sample-code filename-stem linkage.
#   3. Extract embedded PDF text where available (NO OCR).
#   4. Inspect ISPD file headers and readable strings without
#      modifying or interpreting the underlying spectra.
#   5. Preserve workbook rows matching each microscopy sample code.
#   6. Write a machine-readable audit and a concise text report.
#
# STRICT BOUNDARIES
#   - No model inference, training, tuning, or threshold changes.
#   - No source files are modified.
#   - Keyword hits are leads for manual review, NOT FTIR results.
#   - Exact sample-code linkage is not object-level ground truth.
# ================================================================

from pathlib import Path
from collections import defaultdict
from datetime import date, datetime
import hashlib
import json
import re
import shutil
import subprocess
import zipfile
import importlib
import traceback

SOURCE_ROOT = Path(
    "/kaggle/input/datasets/abinavprabhuv/dasonet-tioman-source-v1"
)

OUTPUT_DIR = Path(
    "/kaggle/working/dasonet_tioman_ftir_document_content_audit_v1"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

JSON_PATH = OUTPUT_DIR / "TIOMAN_FTIR_DOCUMENT_CONTENT_AUDIT_v1.json"
TXT_PATH = OUTPUT_DIR / "TIOMAN_FTIR_DOCUMENT_CONTENT_AUDIT_v1.txt"

if not SOURCE_ROOT.exists():
    raise FileNotFoundError(
        f"Tioman source mount not found: {SOURCE_ROOT}"
    )


def sha256_file(path, chunk_size=1024 * 1024):
    digest = hashlib.sha256()
    with open(path, "rb") as handle:
        while True:
            chunk = handle.read(chunk_size)
            if not chunk:
                break
            digest.update(chunk)
    return digest.hexdigest()


def relpath(path):
    try:
        return str(path.relative_to(SOURCE_ROOT))
    except Exception:
        return str(path)


def json_safe(value):
    if value is None or isinstance(value, (str, int, float, bool)):
        return value
    if isinstance(value, (datetime, date)):
        return value.isoformat()
    return str(value)


def normalize_text(text):
    return re.sub(r"\s+", " ", text or "").strip()


def index_by_stem(paths):
    result = defaultdict(list)
    for path in paths:
        result[path.stem].append(path)
    return result


def printable_ascii_strings(raw, minimum_length=4):
    # A limited, read-only printable-string scan. This is not a parser.
    found = re.findall(rb"[\x20-\x7e]{" + str(minimum_length).encode() + rb",}", raw)
    return [
        item.decode("ascii", errors="replace")
        for item in found
    ]


# These are discovery terms only. A hit does not establish a polymer ID.
TERM_PATTERNS = {
    "polyethylene": r"\bpolyethylene\b",
    "polypropylene": r"\bpolypropylene\b",
    "polystyrene": r"\bpolystyrene\b",
    "polyethylene_terephthalate": r"\bpolyethylene\s+terephthalate\b",
    "polyvinyl_chloride": r"\bpolyvinyl\s+chloride\b",
    "polyamide": r"\bpolyamide\b",
    "nylon": r"\bnylon\b",
    "polyurethane": r"\bpolyurethane\b",
    "polycarbonate": r"\bpolycarbonate\b",
    "polymethyl_methacrylate": r"\bpolymethyl\s+methacrylate\b",
    "polytetrafluoroethylene": r"\bpolytetrafluoroethylene\b",
    "polyvinylidene_fluoride": r"\bpolyvinylidene\s+fluoride\b",
    "polyoxymethylene": r"\bpolyoxymethylene\b",
    "polymer": r"\bpolymers?\b",
    "plastic": r"\bplastics?\b",
    "microplastic": r"\bmicroplastics?\b",
    "ftir": r"\bFT[\s-]?IR\b",
    "spectrum": r"\bspectr(?:um|a)\b",
    "library_match": r"\blibrary\s+(?:match|search|hit|result)\b",
    "match_score": r"\b(?:match|similarity)\s*(?:score|[%:])\b",
    # Abbreviations are especially prone to false positives. Review context.
    "PET_abbreviation": r"\bPET\b",
    "PE_abbreviation": r"\bPE\b",
    "PP_abbreviation": r"\bPP\b",
    "PVC_abbreviation": r"\bPVC\b",
    "PA_abbreviation": r"\bPA\b",
    "ABS_abbreviation": r"\bABS\b",
    "PMMA_abbreviation": r"\bPMMA\b",
    "PTFE_abbreviation": r"\bPTFE\b",
}


def find_term_hits(text):
    hits = {}
    contexts = {}
    text = text or ""

    for label, pattern in TERM_PATTERNS.items():
        matches = list(re.finditer(pattern, text, flags=re.IGNORECASE))
        if not matches:
            continue

        hits[label] = len(matches)
        snippets = []
        for match in matches[:3]:
            start = max(0, match.start() - 100)
            end = min(len(text), match.end() + 140)
            snippets.append(normalize_text(text[start:end]))
        contexts[label] = snippets

    return hits, contexts


def extract_pdf_text(path):
    """
    Try pypdf, then PyPDF2, then pdftotext if installed.
    No OCR is attempted.
    """
    errors = []
    text = ""
    page_count = None
    method = None
    metadata = {}

    for module_name in ("pypdf", "PyPDF2"):
        try:
            module = importlib.import_module(module_name)
            reader = module.PdfReader(str(path), strict=False)
            page_count = len(reader.pages)

            page_texts = []
            for page in reader.pages:
                try:
                    page_texts.append(page.extract_text() or "")
                except Exception as exc:
                    page_texts.append("")
                    errors.append(
                        f"{module_name} page extraction: "
                        f"{type(exc).__name__}: {exc}"
                    )

            text = "\n".join(page_texts)
            method = module_name

            try:
                if reader.metadata:
                    metadata = {
                        str(key): str(value)
                        for key, value in reader.metadata.items()
                        if value is not None
                    }
            except Exception:
                pass

            # Prefer pdftotext fallback if the PDF library yielded no text.
            if text.strip():
                break

        except Exception as exc:
            errors.append(
                f"{module_name}: {type(exc).__name__}: {exc}"
            )

    if not text.strip() and shutil.which("pdftotext"):
        try:
            result = subprocess.run(
                ["pdftotext", "-layout", str(path), "-"],
                capture_output=True,
                text=True,
                encoding="utf-8",
                errors="replace",
                timeout=45,
                check=False,
            )
            if result.returncode == 0:
                text = result.stdout or ""
                method = "pdftotext"
            else:
                errors.append(
                    "pdftotext return code "
                    f"{result.returncode}: "
                    f"{(result.stderr or '')[:500]}"
                )
        except Exception as exc:
            errors.append(
                f"pdftotext: {type(exc).__name__}: {exc}"
            )

    if text.strip():
        status = "TEXT_EXTRACTED"
    elif method is not None:
        status = "NO_EMBEDDED_TEXT_FOUND"
    else:
        status = "EXTRACTION_UNAVAILABLE_OR_FAILED"

    return {
        "text": text,
        "page_count": page_count,
        "extraction_method": method,
        "extraction_status": status,
        "metadata": metadata,
        "errors": errors,
    }


def inspect_ispd(path):
    """
    Non-destructive format inspection only.
    This does not decode, identify, or interpret a spectrum.
    """
    record = {
        "path": relpath(path),
        "filename": path.name,
        "stem": path.stem,
        "size_bytes": path.stat().st_size,
        "sha256": sha256_file(path),
    }

    with open(path, "rb") as handle:
        raw = handle.read(4096)

    record["header_first_64_bytes_hex"] = raw[:64].hex()
    record["header_first_64_bytes_ascii"] = "".join(
        chr(byte) if 32 <= byte <= 126 else "."
        for byte in raw[:64]
    )

    try:
        record["is_zip_container"] = zipfile.is_zipfile(path)
    except Exception:
        record["is_zip_container"] = False

    if record["is_zip_container"]:
        try:
            with zipfile.ZipFile(path, "r") as archive:
                record["zip_member_names"] = archive.namelist()[:100]
                record["zip_member_count"] = len(archive.namelist())
        except Exception as exc:
            record["zip_inspection_error"] = (
                f"{type(exc).__name__}: {exc}"
            )

    try:
        with open(path, "rb") as handle:
            raw_for_strings = handle.read(1024 * 1024)

        strings = printable_ascii_strings(raw_for_strings)
        record["printable_string_count_in_first_1MiB"] = len(strings)
        record["printable_strings_preview"] = strings[:30]

        decoded = raw_for_strings.decode("utf-8", errors="replace")
        if raw_for_strings:
            printable_ratio = sum(
                1 for char in decoded
                if char.isprintable() or char in "\r\n\t"
            ) / max(1, len(decoded))
        else:
            printable_ratio = 0.0

        record["utf8_printable_ratio_first_1MiB"] = round(
            printable_ratio, 4
        )

        if printable_ratio >= 0.85:
            record["readable_text_preview"] = decoded[:800]
        else:
            record["readable_text_preview"] = None

    except Exception as exc:
        record["string_scan_error"] = f"{type(exc).__name__}: {exc}"

    record["interpretation_note"] = (
        "Header/string inspection only; no spectral parsing or "
        "polymer identification performed."
    )
    return record


# ------------------------------------------------
# 1. Inventory files
# ------------------------------------------------
all_files = sorted(
    [path for path in SOURCE_ROOT.rglob("*") if path.is_file()],
    key=lambda path: str(path).lower(),
)

jpgs = [path for path in all_files if path.suffix.lower() in (".jpg", ".jpeg")]
pdfs = [path for path in all_files if path.suffix.lower() == ".pdf"]
ispds = [path for path in all_files if path.suffix.lower() == ".ispd"]
workbooks = [
    path for path in all_files
    if path.suffix.lower() == ".xlsx"
    and not path.name.startswith("~$")
]

print("=" * 88)
print("DASO-Net — TIOMAN FTIR DOCUMENT CONTENT AUDIT v1")
print("=" * 88)
print(f"Source root: {SOURCE_ROOT}")
print(f"Total files: {len(all_files)}")
print(f"JPG/JPEG:    {len(jpgs)}")
print(f"PDF:         {len(pdfs)}")
print(f"ISPD:        {len(ispds)}")
print(f"XLSX:        {len(workbooks)}")

if len(jpgs) != 35:
    print(
        "WARNING: Expected 35 microscopy JPG/JPEG files based on the "
        f"previous audit, found {len(jpgs)}."
    )
if len(pdfs) != 35:
    print(
        "WARNING: Expected 35 PDFs based on the previous inventory, "
        f"found {len(pdfs)}."
    )
if len(ispds) != 35:
    print(
        "WARNING: Expected 35 ISPD files based on the previous inventory, "
        f"found {len(ispds)}."
    )

jpg_by_stem = index_by_stem(jpgs)
pdf_by_stem = index_by_stem(pdfs)
ispd_by_stem = index_by_stem(ispds)

# ------------------------------------------------
# 2. Read workbook without altering it
# ------------------------------------------------
workbook_record = {
    "found": False,
    "candidate_paths": [relpath(path) for path in workbooks],
    "sheet_name": None,
    "max_row": None,
    "max_column": None,
    "first_four_rows": [],
    "exact_sample_code_rows": {},
    "errors": [],
}

workbook_code_rows = defaultdict(list)

if workbooks:
    # Prefer the named supporting dataset workbook when present.
    preferred = [
        path for path in workbooks
        if "microparticle sample descriptions" in path.name.lower()
    ]
    workbook_path = preferred[0] if preferred else workbooks[0]

    workbook_record["found"] = True
    workbook_record["path"] = relpath(workbook_path)
    workbook_record["filename"] = workbook_path.name
    workbook_record["size_bytes"] = workbook_path.stat().st_size
    workbook_record["sha256"] = sha256_file(workbook_path)

    try:
        import openpyxl

        wb = openpyxl.load_workbook(
            workbook_path,
            read_only=True,
            data_only=True,
        )
        ws = wb["Sheet1"] if "Sheet1" in wb.sheetnames else wb.active

        workbook_record["sheet_name"] = ws.title
        workbook_record["sheet_names"] = wb.sheetnames
        workbook_record["max_row"] = ws.max_row
        workbook_record["max_column"] = ws.max_column

        for row_number, row in enumerate(
            ws.iter_rows(values_only=True), start=1
        ):
            values = [json_safe(value) for value in row]

            if row_number <= 4:
                workbook_record["first_four_rows"].append({
                    "row_number": row_number,
                    "values": values,
                })

            # Exact cell equality only; no fuzzy code matching.
            for value in values:
                if value is not None:
                    text_value = str(value).strip()
                    if text_value:
                        workbook_code_rows[text_value].append({
                            "row_number": row_number,
                            "values": values,
                        })

        wb.close()

    except Exception as exc:
        workbook_record["errors"].append(
            f"{type(exc).__name__}: {exc}"
        )
        workbook_record["traceback"] = traceback.format_exc()[-3000:]
else:
    workbook_record["errors"].append(
        "No XLSX workbook found under the Tioman source mount."
    )

# ------------------------------------------------
# 3. Hash images and match exact PDF/ISPD stems
# ------------------------------------------------
sample_records = []

for image_path in sorted(jpgs, key=lambda path: path.stem):
    code = image_path.stem

    exact_pdfs = pdf_by_stem.get(code, [])
    exact_ispds = ispd_by_stem.get(code, [])
    matched_workbook_rows = workbook_code_rows.get(code, [])

    sample_records.append({
        "sample_code_from_image_stem": code,
        "image": {
            "path": relpath(image_path),
            "size_bytes": image_path.stat().st_size,
            "sha256": sha256_file(image_path),
        },
        "exact_pdf_stem_match_count": len(exact_pdfs),
        "exact_pdf_paths": [relpath(path) for path in exact_pdfs],
        "exact_ispd_stem_match_count": len(exact_ispds),
        "exact_ispd_paths": [relpath(path) for path in exact_ispds],
        "exact_workbook_code_match_count": len(matched_workbook_rows),
        "workbook_rows": matched_workbook_rows,
    })

# ------------------------------------------------
# 4. Extract PDF text and record candidate terms
# ------------------------------------------------
pdf_records = []

for index, pdf_path in enumerate(
    sorted(pdfs, key=lambda path: path.stem.lower()), start=1
):
    extracted = extract_pdf_text(pdf_path)
    text = extracted["text"] or ""
    hits, contexts = find_term_hits(text)

    pdf_records.append({
        "path": relpath(pdf_path),
        "filename": pdf_path.name,
        "stem": pdf_path.stem,
        "size_bytes": pdf_path.stat().st_size,
        "sha256": sha256_file(pdf_path),
        "exact_image_stem_match_count": len(
            jpg_by_stem.get(pdf_path.stem, [])
        ),
        "exact_image_paths": [
            relpath(path) for path in jpg_by_stem.get(pdf_path.stem, [])
        ],
        "page_count": extracted["page_count"],
        "extraction_method": extracted["extraction_method"],
        "extraction_status": extracted["extraction_status"],
        "extracted_character_count": len(text),
        "metadata": extracted["metadata"],
        "candidate_term_hit_counts": hits,
        "candidate_term_contexts": contexts,
        "text_preview": normalize_text(text)[:600],
        # Full embedded text is retained for subsequent source review.
        "extracted_text_full": text,
        "extraction_errors": extracted["errors"],
        "interpretation_note": (
            "Text extraction and keyword discovery only. Candidate terms "
            "are not validated polymer identifications, and no spectra "
            "were interpreted."
        ),
    })

    print(
        f"[PDF {index:02d}/{len(pdfs):02d}] "
        f"{pdf_path.stem} | "
        f"status={extracted['extraction_status']} | "
        f"pages={extracted['page_count']} | "
        f"chars={len(text)} | "
        f"image_stem_match={len(jpg_by_stem.get(pdf_path.stem, []))} | "
        f"terms={','.join(hits.keys()) if hits else '-'}"
    )
    if text.strip():
        print(f"    Preview: {normalize_text(text)[:220]}")
    else:
        print("    Preview: [No embedded text extracted; this is not an OCR result.]")

# ------------------------------------------------
# 5. Inspect ISPD file headers / strings only
# ------------------------------------------------
ispd_records = []

for index, ispd_path in enumerate(
    sorted(ispds, key=lambda path: path.stem.lower()), start=1
):
    record = inspect_ispd(ispd_path)
    record["exact_image_stem_match_count"] = len(
        jpg_by_stem.get(ispd_path.stem, [])
    )
    record["exact_image_paths"] = [
        relpath(path) for path in jpg_by_stem.get(ispd_path.stem, [])
    ]
    ispd_records.append(record)

    strings_preview = record.get("printable_strings_preview", [])
    short_strings = " | ".join(strings_preview[:3])[:180]

    print(
        f"[ISPD {index:02d}/{len(ispds):02d}] "
        f"{ispd_path.stem} | "
        f"bytes={record['size_bytes']} | "
        f"sha256={record['sha256'][:16]}... | "
        f"image_stem_match={record['exact_image_stem_match_count']}"
    )
    if short_strings:
        print(f"    Readable strings (discovery only): {short_strings}")
    else:
        print("    Readable strings: none detected in the inspected bytes")

# ------------------------------------------------
# 6. Summaries / unmatched files
# ------------------------------------------------
image_codes = {path.stem for path in jpgs}
pdf_codes = {path.stem for path in pdfs}
ispd_codes = {path.stem for path in ispds}

missing_pdf_codes = sorted(image_codes - pdf_codes)
missing_ispd_codes = sorted(image_codes - ispd_codes)
unlinked_pdf_stems = sorted(pdf_codes - image_codes)
unlinked_ispd_stems = sorted(ispd_codes - image_codes)

pdf_status_counts = defaultdict(int)
for record in pdf_records:
    pdf_status_counts[record["extraction_status"]] += 1

samples_with_workbook_rows = sum(
    1 for record in sample_records
    if record["exact_workbook_code_match_count"] > 0
)
samples_with_pdf = sum(
    1 for record in sample_records
    if record["exact_pdf_stem_match_count"] > 0
)
samples_with_ispd = sum(
    1 for record in sample_records
    if record["exact_ispd_stem_match_count"] > 0
)

audit = {
    "audit_name": "DASO-Net Tioman FTIR Document Content Audit v1",
    "audit_status": "CONTENT_AUDIT_COMPLETE_REVIEW_REQUIRED",
    "created_utc": datetime.utcnow().isoformat(timespec="seconds") + "Z",
    "source_root": str(SOURCE_ROOT),
    "source_files_read_only": True,
    "model_inference_run": False,
    "training_run": False,
    "threshold_or_decoder_changes": False,
    "ocr_attempted": False,
    "spectral_interpretation_performed": False,
    "polymer_identity_assigned": False,
    "counts": {
        "total_files": len(all_files),
        "jpg_or_jpeg": len(jpgs),
        "pdf": len(pdfs),
        "ispd": len(ispds),
        "xlsx": len(workbooks),
        "images_with_exact_pdf_stem_match": samples_with_pdf,
        "images_with_exact_ispd_stem_match": samples_with_ispd,
        "images_with_exact_workbook_code_match": samples_with_workbook_rows,
        "pdf_extraction_status_counts": dict(pdf_status_counts),
    },
    "filename_stem_linkage": {
        "image_codes_missing_exact_pdf_stem": missing_pdf_codes,
        "image_codes_missing_exact_ispd_stem": missing_ispd_codes,
        "pdf_stems_without_exact_image_stem": unlinked_pdf_stems,
        "ispd_stems_without_exact_image_stem": unlinked_ispd_stems,
        "matching_rule": (
            "Exact case-sensitive Path.stem equality. No fuzzy or inferred "
            "sample-code matching."
        ),
    },
    "workbook": workbook_record,
    "sample_records": sample_records,
    "pdf_documents": pdf_records,
    "ispd_documents": ispd_records,
    "limitations": [
        "Filename linkage alone does not establish a valid FTIR result.",
        "Embedded PDF text may be incomplete or absent; OCR was not run.",
        "ISPD headers and printable strings do not establish file semantics.",
        "Keyword matches are discovery leads, not verified polymer labels.",
        "Workbook sample-level metadata is not object-level bounding-box or mask ground truth.",
        "No inference, training, tuning, or evaluation was performed in this audit.",
    ],
}

# ------------------------------------------------
# 7. Persist complete JSON and concise TXT summary
# ------------------------------------------------
with open(JSON_PATH, "w", encoding="utf-8") as handle:
    json.dump(audit, handle, ensure_ascii=False, indent=2)

json_sha256 = sha256_file(JSON_PATH)

summary_lines = [
    "DASO-Net — TIOMAN FTIR DOCUMENT CONTENT AUDIT v1",
    "=" * 62,
    f"Source root: {SOURCE_ROOT}",
    f"JSON audit: {JSON_PATH}",
    f"JSON SHA-256: {json_sha256}",
    "",
    "COUNTS",
    json.dumps(audit["counts"], indent=2, ensure_ascii=False),
    "",
    "EXACT-STEM LINKAGE",
    json.dumps(audit["filename_stem_linkage"], indent=2, ensure_ascii=False),
    "",
    "PDF EXTRACTION STATUS",
    json.dumps(dict(pdf_status_counts), indent=2, ensure_ascii=False),
    "",
    "PER-SAMPLE SUMMARY",
]

for record in sample_records:
    summary_lines.append(
        f"{record['sample_code_from_image_stem']}: "
        f"PDFs={record['exact_pdf_stem_match_count']}, "
        f"ISPDs={record['exact_ispd_stem_match_count']}, "
        f"workbook_rows={record['exact_workbook_code_match_count']}, "
        f"image_sha256={record['image']['sha256']}"
    )

summary_lines += [
    "",
    "LIMITATIONS / SAFETY BOUNDARIES",
    *[f"- {item}" for item in audit["limitations"]],
    "",
    "No model inference or training was run. Source files were read only.",
]

with open(TXT_PATH, "w", encoding="utf-8") as handle:
    handle.write("\n".join(summary_lines) + "\n")

txt_sha256 = sha256_file(TXT_PATH)

print("\n" + "=" * 88)
print("AUDIT SUMMARY")
print("=" * 88)
print(json.dumps(audit["counts"], indent=2, ensure_ascii=False))
print("\nExact-stem linkage:")
print(json.dumps(audit["filename_stem_linkage"], indent=2, ensure_ascii=False))
print("\nWorkbook SHA-256:")
print(workbook_record.get("sha256", "UNAVAILABLE"))
print("\nSaved files:")
print(f"JSON: {JSON_PATH}")
print(f"JSON SHA-256: {json_sha256}")
print(f"TXT:  {TXT_PATH}")
print(f"TXT SHA-256: {txt_sha256}")
print("\nNo inference, training, tuning, OCR, or spectral interpretation was performed.")
print("NEXT: review extracted document evidence before freezing any Tioman evaluation protocol.")
```

```python

# ================================================================
# DASO-Net — FROZEN EXTERNAL EVALUATOR DISCOVERY AUDIT v1
#
# Read-only discovery. No model loading or inference.
# No edits to project, checkpoint, or source datasets.
# ================================================================

from pathlib import Path
import hashlib
import json
import re
from datetime import datetime, timezone

ROOTS = [
    Path("/kaggle/input/datasets/abinavprabhuv/dasonet-project"),
    Path("/kaggle/input/datasets/abinavprabhuv/dasonet-continuation-checkpoint-v1"),
    Path("/kaggle/working"),
]

OUT_DIR = Path(
    "/kaggle/working/dasonet_frozen_evaluator_discovery_audit_v1"
)
OUT_DIR.mkdir(parents=True, exist_ok=True)

NAME_TERMS = (
    "rsc", "external", "eval", "inference", "protocol",
    "decoder", "frozen", "threshold", "predict",
    "postprocess", "postproc", "recovery", "consistency",
)

TEXT_EXTENSIONS = {
    ".py", ".json", ".yaml", ".yml", ".md", ".txt", ".csv"
}

CONTENT_TERMS = (
    "rsc", "threshold", "decoder", "topk", "nms",
    "preprocess", "checkpoint", "external evaluation",
    "sanitation", "postprocess", "confidence",
)

MAX_TEXT_BYTES = 2_000_000
MAX_CONTENT_HITS_PER_FILE = 12
MAX_PRINT_ROWS = 100


def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()


def safe_read_text(path):
    try:
        return path.read_text(encoding="utf-8", errors="replace")
    except Exception as exc:
        return f""


seen_paths = set()
name_matches = []
content_matches = []
scan_errors = []

for root in ROOTS:
    if not root.exists():
        scan_errors.append({
            "path": str(root),
            "error": "ROOT_NOT_FOUND",
        })
        continue

    for path in root.rglob("*"):
        if not path.is_file():
            continue

        resolved = str(path.resolve())
        if resolved in seen_paths:
            continue
        seen_paths.add(resolved)

        filename_lower = path.name.lower()
        matched_terms = [
            term for term in NAME_TERMS if term in filename_lower
        ]

        if matched_terms:
            try:
                name_matches.append({
                    "path": str(path),
                    "filename": path.name,
                    "size_bytes": path.stat().st_size,
                    "sha256": sha256_file(path),
                    "matched_filename_terms": matched_terms,
                })
            except Exception as exc:
                scan_errors.append({
                    "path": str(path),
                    "error": repr(exc),
                })

        # Content search only for small, text-like files.
        if (
            path.suffix.lower() not in TEXT_EXTENSIONS
            or path.stat().st_size > MAX_TEXT_BYTES
        ):
            continue

        # Avoid scanning the discovery audit's own generated outputs.
        if OUT_DIR in path.parents:
            continue

        text = safe_read_text(path)
        if not text:
            continue

        lower = text.lower()
        matched_content_terms = [
            term for term in CONTENT_TERMS if term in lower
        ]

        if matched_content_terms:
            excerpts = []
            for term in matched_content_terms[:MAX_CONTENT_HITS_PER_FILE]:
                pos = lower.find(term)
                start = max(0, pos - 180)
                end = min(len(text), pos + len(term) + 300)
                excerpts.append({
                    "term": term,
                    "excerpt": re.sub(
                        r"\s+", " ", text[start:end]
                    ).strip(),
                })

            content_matches.append({
                "path": str(path),
                "filename": path.name,
                "size_bytes": path.stat().st_size,
                "sha256": sha256_file(path),
                "matched_content_terms": matched_content_terms,
                "excerpts": excerpts,
            })

name_matches.sort(
    key=lambda x: (
        not ("rsc" in x["filename"].lower()),
        x["filename"].lower(),
        x["path"],
    )
)
content_matches.sort(
    key=lambda x: (
        not ("rsc" in x["filename"].lower()),
        not any(
            term in x["filename"].lower()
            for term in ("eval", "protocol", "inference")
        ),
        x["path"].lower(),
    )
)

report = {
    "audit_name": "DASO-Net Frozen External Evaluator Discovery Audit v1",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "roots": [str(p) for p in ROOTS],
    "read_only": True,
    "model_loaded": False,
    "inference_run": False,
    "training_or_tuning_run": False,
    "name_match_count": len(name_matches),
    "content_match_count": len(content_matches),
    "filename_matches": name_matches,
    "content_matches": content_matches,
    "scan_errors": scan_errors,
    "interpretation_warning": (
        "Search hits are candidates only. The correct evaluator must be "
        "verified against the preserved RSC final run records and exact "
        "checkpoint/decoder/threshold protocol before reuse."
    ),
}

json_path = OUT_DIR / "FROZEN_EXTERNAL_EVALUATOR_DISCOVERY_AUDIT_v1.json"
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2, ensure_ascii=False)

print("=" * 76)
print("DASO-Net — FROZEN EXTERNAL EVALUATOR DISCOVERY AUDIT v1")
print("=" * 76)
print(f"Roots checked: {len(ROOTS)}")
print(f"Unique files traversed: {len(seen_paths)}")
print(f"Filename matches: {len(name_matches)}")
print(f"Text-content matches: {len(content_matches)}")
print(f"Scan errors: {len(scan_errors)}")
print("\nFILENAME MATCHES")
print("-" * 76)

for item in name_matches[:MAX_PRINT_ROWS]:
    print(
        f"{item['filename']} | {item['size_bytes']} bytes | "
        f"SHA256={item['sha256']}"
    )
    print(f"  PATH: {item['path']}")
    print(f"  TERMS: {', '.join(item['matched_filename_terms'])}")

print("\nCONTENT MATCHES — LIKELY RELEVANT FILES")
print("-" * 76)

for item in content_matches[:MAX_PRINT_ROWS]:
    relevant = [
        term for term in item["matched_content_terms"]
        if term in {"rsc", "threshold", "decoder", "topk", "nms",
                    "preprocess", "checkpoint", "external evaluation",
                    "sanitation", "postprocess", "confidence"}
    ]
    if not relevant:
        continue

    print(f"\n{item['filename']} | SHA256={item['sha256']}")
    print(f"  PATH: {item['path']}")
    for excerpt in item["excerpts"][:4]:
        print(f"  [{excerpt['term']}] {excerpt['excerpt'][:360]}")

print("\n" + "=" * 76)
print("AUDIT SAVED")
print(f"JSON: {json_path}")
print(f"SHA256: {sha256_file(json_path)}")
print("No model was loaded; no inference or tuning was performed.")
```

```python

from __future__ import annotations

from pathlib import Path
from collections import Counter
from datetime import datetime, timezone
import csv
import ast
import symtable
import __future__
import builtins
import hashlib
import importlib.util
import json
import math
import os
import sys
import traceback
import gc
import copy

import cv2
import numpy as np
import torch

# ================================================================
# DASO-Net — TIOMAN FROZEN THREE-SEED INFERENCE RESUME03
#
# Complete standalone Kaggle notebook cell.
#
# FIX:
#   The frozen RSC checkpoint loader requires pathlib.Path.
#   Previous Resume02 passed str(ckpt_path), causing AttributeError.
#
# FROZEN:
#   Model, decoder, checkpoints, thresholds, source images,
#   tile schedule and qualitative evaluation protocol unchanged.
#
# NO:
#   Training, tuning, calibration, seed selection, RSC rerun,
#   ground-truth-based metrics, or alteration of frozen inputs.
#
# Each attempt creates a fresh output directory.
# ================================================================

CONT = Path(
    "/kaggle/input/datasets/abinavprabhuv/"
    "dasonet-continuation-checkpoint-v1/"
    "DASO_NET_CONTINUATION_CHECKPOINT_v1/"
    "dasonet_continuation_checkpoint_v1"
)

BUNDLE = CONT / "bundle"
CODE = BUNDLE / "phase5b_code"
SOURCE_CODE = CODE / "src"

TIOMAN = Path(
    "/kaggle/input/datasets/abinavprabhuv/"
    "dasonet-tioman-source-v1"
)

RSC = (
    BUNDLE
    / "rsc_final/RSC_FORMAL_RECOVERY_RESULTS_FREEZE_v1/"
    "rsc_formal_recovery_results_freeze_v1/"
    "preserved_artifacts"
)

FREEZE_DIR = Path(
    "/kaggle/working/"
    "dasonet_tioman_qualitative_tiling_protocol_freeze_v1"
)

PROTOCOL_FILE = (
    FREEZE_DIR
    / "TIOMAN_QUALITATIVE_TILING_PROTOCOL_FREEZE_v1.json"
)

SCHEDULE_FILE = (
    FREEZE_DIR
    / "TIOMAN_TILE_SCHEDULE_v1.csv"
)

RESUME02_FILE = Path(
    "/kaggle/working/"
    "dasonet_tioman_preprotocol_interface_audit_v1_resume02/"
    "TIOMAN_PREPROTOCOL_INTERFACE_AUDIT_RESUME02.json"
)

CAPTURE_FILE = Path(
    "/kaggle/working/"
    "dasonet_tioman_execution_interface_capture_v1/"
    "TIOMAN_FROZEN_EXECUTION_INTERFACE_CAPTURE_v1.json"
)

OUT_BASE = Path(
    "/kaggle/working/"
    "dasonet_tioman_frozen_three_seed_inference_v1"
)

# ================================================================
# EXACT FROZEN IDENTITIES
# ================================================================

EXPECTED = {
    "protocol": (
        PROTOCOL_FILE,
        "75b81b88550d7f76f6e7cecf27c4e805a281e350a99159ebba8c4447daaae563",
    ),
    "schedule": (
        SCHEDULE_FILE,
        "4aa24b8864052108146d4df9e1db490569bed1172f8b12e78aa95260b1c70f7f",
    ),
    "resume02": (
        RESUME02_FILE,
        "ca505ee2597dcf75adb575ec81e64d7918fb4ec2edf1f24583ab96ea0a938e91",
    ),
    "interface_capture": (
        CAPTURE_FILE,
        "20ad2da7a61fed515ea9f565791a20b4931c92859d59e0917c5473dc9e3403a0",
    ),
    "rsc_original": (
        RSC
        / "05_original_evaluator_implementation_freeze/"
        "evaluate_rsc_external_v1_FROZEN.py",
        "59a3de31f2a5244ba32c53b705b050e033b5d17c8de7efb36c17dbc49f0a7723",
    ),
    "rsc_recovery": (
        RSC
        / "12_recovery_evaluator_implementation_freeze/"
        "evaluate_rsc_external_recovery_v1_FROZEN.py",
        "d0df4ad7cdd60aae23fdff7d2c708c938952510df4a8d9d7a9087eecbad85bda",
    ),
    "phase5c_eval": (
        CODE / "scripts/evaluate_phase5c.py",
        "9c6221b081121f34c0008c1ff49205758d5e551c01641b52de7643e5cf7e4e14",
    ),
    "model_definition": (
        SOURCE_CODE / "dasonet/model/dasonet.py",
        "1ac6f92e1683c6939ff5783ac0cae065dab20e366dd49a9b72ac89007e5eaa40",
    ),
    "decoder": (
        SOURCE_CODE / "dasonet/model/decoding.py",
        "9799eb1be568e3d6e0ce82b7c99df7a0ed05d271cd44ff24b612526cc18537d1",
    ),
    "rsc_scientific_protocol": (
        RSC
        / "02_protocol_freeze/"
        "RSC_EXTERNAL_EVALUATION_PROTOCOL_FREEZE_v1.json",
        "e3c87caebafe87c7671b5745a480c0edf67f444feac4278633da6addae4d1f7e",
    ),
    "rsc_recovery_policy": (
        RSC
        / "11_recovery_policy_freeze/"
        "RSC_DEGENERATE_BOX_RECOVERY_POLICY_FREEZE_v1.json",
        "aa754df77f897c68421a065ffa0b0ffcfe6b332b8dba96b6415d089cb2216f8b",
    ),
}

EXPECTED_SEEDS = [
    (
        1,
        1782656963,
        0.30,
        "seed01_s1782656963_opt002750.pt",
        "e4c4b79ad4b8c6e18ba2b1e3d6cbe796c9cc301bd6fcc7c596fc290ad3a7e81d",
    ),
    (
        2,
        901395089,
        0.35,
        "seed02_s901395089_opt002750.pt",
        "19cdb834a46240916b2c8023e8c3afe5738d9c7cd5d8f818c62ad1e289df1ad2",
    ),
    (
        3,
        252916394,
        0.35,
        "seed03_s252916394_opt002750.pt",
        "be7b7941f2f62179e15074aa973bf6da0eb9018ec9133486512bd1f87153f447",
    ),
]

FROZEN_DECODER = {
    "stride": 4,
    "decode_min_score": 0.05,
    "topk": 100,
    "nms_iou": 0.5,
    "mask_threshold": 0.5,
    "local_peak_kernel": 3,
}

# ================================================================
# GENERAL HELPERS
# ================================================================

def require(condition, message):
    if not condition:
        raise RuntimeError(message)


def sha256_file(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for chunk in iter(
            lambda: f.read(4 * 1024 * 1024),
            b"",
        ):
            h.update(chunk)
    return h.hexdigest()


def verify(path, digest, name):
    require(
        path.is_file(),
        f"Missing {name}: {path}",
    )
    actual = sha256_file(path)
    require(
        actual == digest,
        f"FROZEN SHA MISMATCH {name}: "
        f"expected={digest}, actual={actual}",
    )
    return str(path)


def load_json(path):
    return json.loads(
        Path(path).read_text(encoding="utf-8")
    )


def iso_now():
    return datetime.now(timezone.utc).isoformat()


def fresh_output_dir(base):
    for n in range(100):
        path = (
            base if n == 0
            else Path(str(base) + f"_resume{n:02d}")
        )
        if not path.exists():
            path.mkdir(
                parents=True,
                exist_ok=False,
            )
            return path

    raise RuntimeError(
        "No unused run folder; existing evidence "
        "must not be overwritten."
    )


def atomic_json(path, data):
    path = Path(path)
    tmp = path.with_name(path.name + ".tmp")

    with tmp.open("w", encoding="utf-8") as f:
        json.dump(
            data,
            f,
            sort_keys=True,
            indent=2,
            ensure_ascii=False,
            allow_nan=False,
        )
        f.write("\n")
        f.flush()
        os.fsync(f.fileno())

    os.replace(tmp, path)


def durable_jsonl_append(handle, item):
    serialized = json.dumps(
        item,
        sort_keys=True,
        ensure_ascii=False,
        allow_nan=False,
        separators=(",", ":"),
    )
    handle.write(serialized + "\n")
    handle.flush()
    os.fsync(handle.fileno())


def load_verified_module(alias, path):
    spec = importlib.util.spec_from_file_location(
        alias,
        str(path),
    )
    require(
        spec is not None and spec.loader is not None,
        f"Cannot build module spec: {alias}",
    )

    module = importlib.util.module_from_spec(spec)
    sys.modules[alias] = module
    spec.loader.exec_module(module)
    return module

# ================================================================
# RESTORE EXACT FROZEN RECOVERY DIAGNOSTIC INITIALIZER
# ================================================================

def frozen_diagnostic_initializer(
    frozen_tree,
    frozen_text,
    name,
):
    """
    Restore the original module-level diagnostic initializer.

    The frozen recovery module cannot be imported normally
    because it eagerly references an obsolete Kaggle working
    path. The initializer is extracted without guessing its
    type and without running module startup code.
    """
    nodes = []

    for node in frozen_tree.body:
        if isinstance(node, ast.Assign):
            if any(
                isinstance(target, ast.Name)
                and target.id == name
                for target in node.targets
            ):
                require(
                    len(node.targets) == 1
                    and isinstance(
                        node.targets[0],
                        ast.Name,
                    ),
                    f"Ambiguous initializer for {name}",
                )
                nodes.append((node, node.value))

        elif isinstance(node, ast.AnnAssign):
            if (
                isinstance(node.target, ast.Name)
                and node.target.id == name
            ):
                nodes.append((node, node.value))

        elif isinstance(node, ast.AugAssign):
            if (
                isinstance(node.target, ast.Name)
                and node.target.id == name
            ):
                raise RuntimeError(
                    f"Unsafe augmented initializer for {name}"
                )

    require(
        len(nodes) == 1
        and nodes[0][1] is not None,
        f"Expected one frozen initializer for {name}",
    )

    statement, expression = nodes[0]

    init_text = (
        ast.get_source_segment(
            frozen_text,
            statement,
        )
        or ast.unparse(statement)
    )

    def safe_value(node):
        try:
            return ast.literal_eval(node)
        except (TypeError, ValueError, RecursionError):
            pass

        require(
            isinstance(node, ast.Call)
            and isinstance(node.func, ast.Name)
            and node.func.id in {
                "dict",
                "list",
                "tuple",
                "set",
                "Counter",
            }
            and not any(
                keyword.arg is None
                for keyword in node.keywords
            ),
            "Unsupported frozen initializer: "
            + ast.unparse(node),
        )

        args = [
            safe_value(arg)
            for arg in node.args
        ]
        kwargs = {
            keyword.arg: safe_value(keyword.value)
            for keyword in node.keywords
        }

        constructors = {
            "dict": dict,
            "list": list,
            "tuple": tuple,
            "set": set,
            "Counter": Counter,
        }

        return constructors[node.func.id](
            *args,
            **kwargs,
        )

    obj = safe_value(expression)

    require(
        isinstance(
            obj,
            (
                dict,
                list,
                tuple,
                set,
                Counter,
                type(None),
            ),
        ),
        f"Unexpected diagnostic type: {type(obj).__name__}",
    )

    return obj, {
        "source_line": statement.lineno,
        "source_expression": init_text[:1200],
        "container_type": type(obj).__name__,
    }


def load_frozen_recovery_geometry_without_module_side_effects(
    path,
    original,
    verified_policy_path,
):
    """
    Compile the exact frozen recovery function from the
    SHA-verified original source.

    No frozen source file is altered.
    All unresolved dependencies remain fatal.
    """
    frozen_text = Path(path).read_text(
        encoding="utf-8"
    )
    tree = ast.parse(
        frozen_text,
        filename=str(path),
    )

    function_name = (
        "recovery_apply_original_image_box_policy"
    )

    matches = [
        node
        for node in tree.body
        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef),
        )
        and node.name == function_name
    ]

    require(
        len(matches) == 1
        and isinstance(
            matches[0],
            ast.FunctionDef,
        ),
        "Frozen recovery geometry function "
        "must be uniquely defined.",
    )

    node = matches[0]

    require(
        not node.decorator_list,
        "Unexpected decorator on frozen recovery geometry.",
    )

    policy_sha = sha256_file(verified_policy_path)

    require(
        policy_sha == EXPECTED["rsc_recovery_policy"][1],
        "Frozen recovery policy SHA changed.",
    )

    frozen_policy = load_json(
        verified_policy_path
    )

    diagnostics, diagnostic_audit = (
        frozen_diagnostic_initializer(
            tree,
            frozen_text,
            "RECOVERY_CALL_DIAGNOSTICS",
        )
    )

    namespace = {
        "__builtins__": builtins.__dict__,
        "__name__": (
            "dasonet_tioman_frozen_"
            "recovery_geometry_ast_resume03"
        ),
        "np": np,
        "numpy": np,
        "math": math,
        "torch": torch,
        "json": json,
        "re": __import__("re"),
        "Path": Path,
        "Counter": Counter,
        "require": (
            original.require
            if callable(
                getattr(original, "require", None)
            )
            else require
        ),
        "RECOVERY_POLICY_OBJECT": frozen_policy,
        "RECOVERY_POLICY_SHA256": policy_sha,
        "RECOVERY_POLICY": verified_policy_path,
        "RECOVERY_CALL_DIAGNOSTICS": diagnostics,
    }

    symbols = symtable.symtable(
        frozen_text,
        str(path),
        "exec",
    )

    scopes = [
        scope
        for scope in symbols.get_children()
        if scope.get_name() == function_name
    ]

    require(
        len(scopes) == 1,
        "Frozen recovery function symbol scope unavailable.",
    )

    globals_used = set()

    def gather(scope):
        for symbol in scope.get_symbols():
            if (
                symbol.is_global()
                and symbol.is_referenced()
            ):
                globals_used.add(
                    symbol.get_name()
                )

        for child in scope.get_children():
            gather(child)

    gather(scopes[0])

    unbound = sorted(
        name
        for name in globals_used
        if name not in namespace
        and not hasattr(builtins, name)
    )

    print(
        "FROZEN RECOVERY FUNCTION GLOBALS:",
        sorted(globals_used),
        flush=True,
    )
    print(
        "RECOVERY_CALL_DIAGNOSTICS restored:",
        f"line={diagnostic_audit['source_line']}",
        f"type={diagnostic_audit['container_type']}",
        flush=True,
    )

    require(
        not unbound,
        "Unresolved globals in frozen recovery geometry: "
        + ", ".join(unbound),
    )

    future_annotations = any(
        isinstance(item, ast.ImportFrom)
        and item.module == "__future__"
        and any(
            alias.name == "annotations"
            for alias in item.names
        )
        for item in tree.body
    )

    flags = (
        __future__.annotations.compiler_flag
        if future_annotations
        else 0
    )

    isolated = ast.fix_missing_locations(
        ast.Module(
            body=[node],
            type_ignores=[],
        )
    )

    exec(
        compile(
            isolated,
            str(path),
            "exec",
            flags=flags,
            dont_inherit=True,
        ),
        namespace,
    )

    recovered = namespace.get(function_name)

    require(
        callable(recovered),
        "Frozen recovery geometry failed to compile.",
    )

    audit = {
        "source_sha256": sha256_file(path),
        "source_function_lines": [
            node.lineno,
            node.end_lineno,
        ],
        "globals_used": sorted(globals_used),
        "diagnostic_initializer": diagnostic_audit,
        "policy_sha256": policy_sha,
        "technique": (
            "Unchanged frozen AST function with exact "
            "module-level diagnostic initializer; "
            "no recovery module startup."
        ),
    }

    return recovered, audit

# ================================================================
# FROZEN TILE SCHEDULE
# ================================================================

def rect_from_schedule(row):
    return {
        key: int(row[key])
        for key in (
            "tile_x0",
            "tile_y0",
            "tile_x1",
            "tile_y1",
            "owner_x0",
            "owner_y0",
            "owner_x1",
            "owner_y1",
            "row_index",
            "column_index",
            "tile_size",
        )
    }


def load_and_validate_schedule(protocol):
    with SCHEDULE_FILE.open(
        "r",
        newline="",
        encoding="utf-8",
    ) as f:
        rows = list(csv.DictReader(f))

    require(
        len(rows) == 420,
        "Frozen tile schedule row count is not 420.",
    )

    require(
        len({
            row["tile_id"]
            for row in rows
        }) == 420,
        "Duplicate frozen tile ID.",
    )

    images = protocol["images"]

    image_by_code = {
        item["sample_code"]: item
        for item in images
    }

    require(
        len(image_by_code) == 35
        and len(images) == 35,
        "Frozen image population differs from 35.",
    )

    geometry = protocol["spatial_tiling"]

    xs = geometry["tile_x_origins"]
    ys = geometry["tile_y_origins"]
    xe = geometry["owner_x_edges"]
    ye = geometry["owner_y_edges"]

    require(
        xs == [0, 939, 1877, 2816]
        and ys == [0, 568, 1136],
        "Frozen tile origins changed.",
    )

    require(
        xe == [0, 981, 1920, 2858, 3840]
        and ye == [0, 796, 1364, 2160],
        "Frozen ownership edges changed.",
    )

    require(
        geometry["tile_width"] == 1024
        and geometry["tile_height"] == 1024,
        "Frozen tile size changed.",
    )

    expected = []

    for code in sorted(image_by_code):
        image = image_by_code[code]

        for ri, y in enumerate(ys):
            for ci, x in enumerate(xs):
                expected.append(
                    (code, ri, ci, x, y, image)
                )

    require(
        len(expected) == len(rows),
        "Frozen schedule size mismatch.",
    )

    for index, (row, spec) in enumerate(
        zip(rows, expected)
    ):
        code, ri, ci, x, y, image = spec

        fields = rect_from_schedule(row)

        required = {
            "tile_x0": x,
            "tile_y0": y,
            "tile_x1": x + 1024,
            "tile_y1": y + 1024,
            "owner_x0": xe[ci],
            "owner_y0": ye[ri],
            "owner_x1": xe[ci + 1],
            "owner_y1": ye[ri + 1],
            "row_index": ri,
            "column_index": ci,
            "tile_size": 1024,
        }

        require(
            fields == required,
            f"Frozen tile geometry differs at row {index + 1}.",
        )

        require(
            row["sample_code"] == code,
            "Frozen schedule image order changed.",
        )

        require(
            row["tile_id"]
            == f"{code}__r{ri:02d}_c{ci:02d}",
            "Frozen tile ID changed.",
        )

        require(
            row["source_image_path"]
            == image["image_path"],
            "Schedule source path changed.",
        )

        require(
            row["original_image_sha256"]
            == image["image_sha256"],
            "Schedule image SHA changed.",
        )

        require(
            xe[ci] >= x
            and xe[ci + 1] <= x + 1024,
            "Ownership exceeds source tile horizontally.",
        )

        require(
            ye[ri] >= y
            and ye[ri + 1] <= y + 1024,
            "Ownership exceeds source tile vertically.",
        )

    return rows


def image_source_preflight(protocol):
    for item in protocol["images"]:
        path = Path(item["image_path"])

        require(
            path.is_file(),
            f"Source JPEG missing: {path}",
        )

        require(
            path.resolve().is_relative_to(
                TIOMAN.resolve()
            ),
            "Source image outside Tioman mount.",
        )

        require(
            sha256_file(path)
            == item["image_sha256"],
            f"Frozen image hash mismatch: {item['sample_code']}",
        )

        require(
            item["width"] == 3840
            and item["height"] == 2160,
            "Frozen source dimensions changed.",
        )

        image = cv2.imread(
            str(path),
            cv2.IMREAD_COLOR,
        )

        require(
            image is not None
            and image.shape == (2160, 3840, 3),
            f"Cannot decode original JPEG: {path}",
        )

    print(
        "PASS 35/35 frozen microscopy source images.",
        flush=True,
    )

# ================================================================
# ORIGINAL-IMAGE COORDINATE RECONSTRUCTION
# ================================================================

def prediction_record(
    prediction,
    tile,
    threshold,
):
    require(
        isinstance(prediction, dict),
        "Unexpected normalized decoder instance type.",
    )

    box = prediction.get("box")

    require(
        box is not None and len(box) == 4,
        "Decoded prediction box not length four.",
    )

    local_box = [float(value) for value in box]
    score = float(prediction["score"])

    require(
        all(math.isfinite(value) for value in local_box)
        and math.isfinite(score),
        "Non-finite prediction under frozen recovery policy.",
    )

    require(
        0.05 - 1e-8 <= score <= 1.0 + 1e-8,
        "Prediction violates frozen decoder floor.",
    )

    require(
        local_box[2] > local_box[0]
        and local_box[3] > local_box[1],
        "Degenerate box survived frozen recovery.",
    )

    require(
        local_box[0] >= -1e-6
        and local_box[1] >= -1e-6
        and local_box[2] <= 1024 + 1e-6
        and local_box[3] <= 1024 + 1e-6,
        "Recovered box extends beyond 1024-pixel tile.",
    )

    tx = int(tile["tile_x0"])
    ty = int(tile["tile_y0"])

    image_box = [
        local_box[0] + tx,
        local_box[1] + ty,
        local_box[2] + tx,
        local_box[3] + ty,
    ]

    cx = (image_box[0] + image_box[2]) / 2.0
    cy = (image_box[1] + image_box[3]) / 2.0

    owned = (
        int(tile["owner_x0"])
        <= cx
        < int(tile["owner_x1"])
        and int(tile["owner_y0"])
        <= cy
        < int(tile["owner_y1"])
    )

    return {
        "tile_id": tile["tile_id"],
        "local_box_xyxy": local_box,
        "image_box_xyxy": image_box,
        "score": score,
        "decoder_order": int(
            prediction.get("decoder_order", 0)
        ),
        "original_center_xy": [cx, cy],
        "within_frozen_ownership_region": bool(owned),
        "passes_frozen_operating_threshold": bool(
            score >= threshold
        ),
    }


def draw_overlay(
    image_path,
    predictions,
    code,
    seed_num,
    threshold,
    output_path,
):
    source = cv2.imread(
        str(image_path),
        cv2.IMREAD_COLOR,
    )

    require(
        source is not None
        and source.shape == (2160, 3840, 3),
        "Overlay source JPEG decode failed.",
    )

    for row in predictions:
        x1, y1, x2, y2 = row["image_box_xyxy"]

        p1 = (
            max(0, min(3839, int(round(x1)))),
            max(0, min(2159, int(round(y1)))),
        )
        p2 = (
            max(0, min(3839, int(round(x2)))),
            max(0, min(2159, int(round(y2)))),
        )

        cv2.rectangle(
            source,
            p1,
            p2,
            (0, 220, 255),
            3,
        )

        cv2.putText(
            source,
            f"{row['score']:.2f}",
            (p1[0], max(15, p1[1] - 5)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.55,
            (0, 220, 255),
            2,
            cv2.LINE_AA,
        )

    header = (
        f"Tioman external qualitative | {code} | "
        f"Seed-{seed_num} | threshold={threshold:.2f} | "
        f"n={len(predictions)}"
    )

    cv2.rectangle(
        source,
        (0, 0),
        (3839, 53),
        (0, 0, 0),
        -1,
    )

    cv2.putText(
        source,
        header,
        (16, 35),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.92,
        (255, 255, 255),
        2,
        cv2.LINE_AA,
    )

    require(
        cv2.imwrite(
            str(output_path),
            source,
            [cv2.IMWRITE_JPEG_QUALITY, 90],
        ),
        f"Failed to write overlay: {output_path}",
    )

    return sha256_file(output_path)

# ================================================================
# NEW RUN DIRECTORY AND INITIAL STATE
# ================================================================

run_dir = fresh_output_dir(OUT_BASE)

state_path = (
    run_dir
    / "TIOMAN_THREE_SEED_RUN_STATE_v1.json"
)

state = {
    "run_name": (
        "TIOMAN_FROZEN_THREE_SEED_"
        "QUALITATIVE_INFERENCE_v1_RESUME03"
    ),
    "created_utc": iso_now(),
    "status": "PREFLIGHT_NOT_YET_COMPLETE",
    "run_directory": str(run_dir),
    "all_three_seeds_planned": True,
    "frozen_expected_tile_forwards": 1260,
    "completed_tile_forwards": 0,
    "completed_images": 0,
    "model_retraining": False,
    "calibration_or_tuning": False,
    "external_GT_based_metrics": False,
    "rsc_rerun": False,
    "frozen_protocol_sha256": EXPECTED["protocol"][1],
    "frozen_schedule_sha256": EXPECTED["schedule"][1],
    "historical_derived_audit_json_byte_mismatches_preserved": True,
    "seed_results": {},
    "code_and_input_hashes": {},
}

atomic_json(state_path, state)

print("=" * 82)
print(
    "DASO-Net — TIOMAN FROZEN THREE-SEED "
    "QUALITATIVE INFERENCE RESUME03"
)
print("NEW OUTPUT DIRECTORY:", run_dir)
print(
    "FROZEN PROTOCOL SHA256:",
    EXPECTED["protocol"][1],
)
print(
    "FROZEN TILE SCHEDULE SHA256:",
    EXPECTED["schedule"][1],
)
print(
    "NO training, tuning, decoder change, threshold "
    "search, RSC rerun, or GT metrics."
)
print("=" * 82)

try:
    # ============================================================
    # A. EXACT FROZEN INPUTS AND PROTOCOL
    # ============================================================

    print("\nA. EXACT FROZEN INPUTS AND PROTOCOL")

    for label, (path, digest) in EXPECTED.items():
        verify(path, digest, label)

        state["code_and_input_hashes"][label] = {
            "path": str(path),
            "sha256": digest,
        }

        print(
            "PASS",
            label,
            digest[:16] + "...",
        )

    protocol = load_json(PROTOCOL_FILE)

    require(
        protocol["protocol_status"]
        == "FROZEN_BEFORE_TIOMAN_MODEL_INFERENCE",
        "Tioman protocol is not frozen.",
    )

    require(
        protocol["tile_schedule_csv"]["sha256"]
        == EXPECTED["schedule"][1]
        and protocol["tile_schedule_csv"]["records"] == 420,
        "Protocol references another tile schedule.",
    )

    require(
        protocol["spatial_tiling"]["global_cross_tile_nms"]
        == (
            "NONE. Frozen per-tile decoder NMS remains "
            "enabled; no newly tuned global NMS. "
            "Residual boundary duplicates acknowledged."
        ),
        "Global suppression protocol changed.",
    )

    require(
        protocol["spatial_tiling"]["ownership_filter"].startswith(
            "Keep a candidate only if"
        ),
        "Ownership filter protocol changed.",
    )

    require(
        protocol["reported_outputs"][
            "save_raw_floor_predictions"
        ] is True
        and protocol["reported_outputs"][
            "save_thresholded_core_filtered_predictions"
        ] is True,
        "Frozen output requirements changed.",
    )

    require(
        protocol["restrictions"][
            "all_three_frozen_seeds_mandatory"
        ] is True
        and protocol["restrictions"][
            "no_tioman_training"
        ] is True
        and protocol["restrictions"][
            "no_threshold_tuning"
        ] is True,
        "Frozen restrictions changed.",
    )

    decoder_spec = protocol["decoder"]

    require(
        decoder_spec["score_threshold_floor"] == 0.05
        and decoder_spec["stride"] == 4
        and decoder_spec["topk_per_tile"] == 100
        and decoder_spec["nms_iou_per_tile"] == 0.5
        and decoder_spec["mask_threshold"] == 0.5
        and decoder_spec["local_peak_kernel"] == 3,
        "Frozen decoder settings changed.",
    )

    require(
        protocol["source_provenance"]["resume02"]["sha256"]
        == EXPECTED["resume02"][1],
        "Resume02 evidence chain changed.",
    )

    capture = load_json(CAPTURE_FILE)

    require(
        capture["status"]
        == "SOURCE_INTERFACE_CAPTURED_NO_INFERENCE",
        "Execution-interface capture is not PASS.",
    )

    require(
        len(protocol["final_seeds"]) == 3,
        "Protocol does not specify three seeds.",
    )

    for frozen, spec in zip(
        protocol["final_seeds"],
        EXPECTED_SEEDS,
    ):
        index, seed, threshold, filename, digest = spec

        require(
            frozen["index"] == index
            and frozen["training_seed"] == seed
            and frozen["frozen_score_threshold"] == threshold
            and frozen["optimizer_updates"] == 2750
            and frozen["checkpoint"]["sha256"] == digest,
            f"Frozen Seed-{index} differs from protocol.",
        )

        verify(
            BUNDLE / "final_checkpoints" / filename,
            digest,
            f"Seed-{index} checkpoint",
        )

    schedule = load_and_validate_schedule(protocol)
    image_source_preflight(protocol)

    all_python_sources = sorted(
        path
        for path in SOURCE_CODE.rglob("*.py")
        if path.is_file()
    )

    state["portable_package_python_sha256"] = {
        str(path.relative_to(SOURCE_CODE)): sha256_file(path)
        for path in all_python_sources
    }

    state["status"] = "PREFLIGHT_PASSED"
    atomic_json(state_path, state)

    print(
        "PASS 420 tiles; 35 source images; "
        "3 frozen checkpoints."
    )

    # ============================================================
    # B. UNCHANGED FROZEN EVALUATOR FUNCTIONS
    # ============================================================

    print(
        "\nB. LOADING VERIFIED FROZEN EVALUATOR FUNCTIONS"
    )

    source_root = SOURCE_CODE.resolve()

    for name, module in list(sys.modules.items()):
        if (
            name == "dasonet"
            or name.startswith("dasonet.")
        ):
            location = getattr(
                module,
                "__file__",
                None,
            )

            require(
                location is not None
                and Path(location).resolve().is_relative_to(
                    source_root
                ),
                f"Conflicting dasonet module: "
                f"{name} => {location}",
            )

    if str(SOURCE_CODE) not in sys.path:
        sys.path.insert(
            0,
            str(SOURCE_CODE),
        )

    original = load_verified_module(
        "dasonet_tioman_rsc_original_resume03",
        EXPECTED["rsc_original"][0],
    )

    recovery_geometry, recovery_audit = (
        load_frozen_recovery_geometry_without_module_side_effects(
            EXPECTED["rsc_recovery"][0],
            original,
            EXPECTED["rsc_recovery_policy"][0],
        )
    )

    state["recovery_loader_audit"] = recovery_audit
    atomic_json(state_path, state)

    phase5c = load_verified_module(
        "dasonet_tioman_phase5c_resume03",
        EXPECTED["phase5c_eval"][0],
    )

    from dasonet.model.decoding import decode_instances
    import dasonet.model.decoding as decoder_module

    require(
        Path(decoder_module.__file__).resolve()
        == EXPECTED["decoder"][0].resolve(),
        "Imported decoder source path differs.",
    )

    require(
        sha256_file(decoder_module.__file__)
        == EXPECTED["decoder"][1],
        "Imported decoder source hash differs.",
    )

    for name, function in [
        (
            "frozen RSC preprocessor",
            original.preprocess_rsc_image,
        ),
        (
            "frozen decoder normalization",
            original.normalize_decoded_instances,
        ),
        (
            "frozen RSC infer_one",
            original.infer_one,
        ),
        (
            "frozen recovery geometry",
            recovery_geometry,
        ),
        (
            "frozen checkpoint loader",
            original.load_checkpoint_mapping,
        ),
        (
            "frozen Phase5C model builder",
            phase5c.build_model,
        ),
    ]:
        require(
            callable(function),
            f"Missing frozen callable: {name}",
        )

    # Delegate to the byte-frozen recovery geometry function.
    # This modifies only the in-memory evaluator namespace,
    # never its source file or the model/decoder.
    original.apply_original_image_box_policy = (
        recovery_geometry
    )

    valid_smoke = recovery_geometry(
        [
            {
                "box": [25.0, 35.0, 45.0, 70.0],
                "score": 0.8,
                "decoder_order": 0,
            }
        ],
        1024,
        1024,
    )

    require(
        isinstance(valid_smoke, list)
        and len(valid_smoke) == 1,
        "Frozen recovery valid-box self-test failed.",
    )

    degenerate_smoke = recovery_geometry(
        [
            {
                "box": [25.0, 35.0, 25.0, 70.0],
                "score": 0.8,
                "decoder_order": 0,
            }
        ],
        1024,
        1024,
    )

    require(
        isinstance(degenerate_smoke, list)
        and len(degenerate_smoke) == 0,
        "Frozen recovery zero-area self-test failed.",
    )

    # Discard synthetic-test diagnostics and restore the exact
    # initial diagnostic state before scientific inference.
    frozen_recovery_source = (
        EXPECTED["rsc_recovery"][0].read_text(
            encoding="utf-8"
        )
    )

    clean_diagnostics, _ = frozen_diagnostic_initializer(
        ast.parse(frozen_recovery_source),
        frozen_recovery_source,
        "RECOVERY_CALL_DIAGNOSTICS",
    )

    recovery_geometry.__globals__[
        "RECOVERY_CALL_DIAGNOSTICS"
    ] = copy.deepcopy(clean_diagnostics)

    print(
        "PASS frozen recovery geometry and self-tests."
    )
    print(
        "PASS original diagnostic state restored."
    )

    # ============================================================
    # C. LOSSLESS TILE BRIDGE AND PREPROCESSING PARITY
    # ============================================================

    bridge_path = (
        run_dir
        / "tioman_frozen_tile_bridge_current.png"
    )

    source_by_code = {
        image["sample_code"]: image
        for image in protocol["images"]
    }

    first_tile = schedule[0]

    first_source = cv2.imread(
        first_tile["source_image_path"],
        cv2.IMREAD_COLOR,
    )

    require(
        first_source is not None
        and first_source.shape == (2160, 3840, 3),
        "Cannot decode first original image.",
    )

    geometry = rect_from_schedule(first_tile)

    first_crop = np.ascontiguousarray(
        first_source[
            geometry["tile_y0"]:geometry["tile_y1"],
            geometry["tile_x0"]:geometry["tile_x1"],
            :,
        ]
    )

    require(
        first_crop.shape == (1024, 1024, 3),
        "First crop dimensions differ.",
    )

    require(
        cv2.imwrite(
            str(bridge_path),
            first_crop,
            [cv2.IMWRITE_PNG_COMPRESSION, 3],
        ),
        "Cannot encode lossless PNG bridge.",
    )

    first_reload = cv2.imread(
        str(bridge_path),
        cv2.IMREAD_COLOR,
    )

    require(
        first_reload is not None
        and np.array_equal(
            first_crop,
            first_reload,
        ),
        "Lossless PNG parity failed.",
    )

    prep_check = original.preprocess_rsc_image(
        str(bridge_path)
    )

    require(
        (
            prep_check["original_w"],
            prep_check["original_h"],
        ) == (1024, 1024)
        and (
            prep_check["padded_w"],
            prep_check["padded_h"],
        ) == (1024, 1024)
        and (
            prep_check["pad_right"],
            prep_check["pad_bottom"],
        ) == (0, 0),
        "Frozen preprocessing geometry differs.",
    )

    expected_array = (
        np.ascontiguousarray(
            first_crop[
                :, :, ::-1
            ].transpose(2, 0, 1)
        ).astype(np.float32)
        / 255.0
    )

    actual_array = (
        prep_check["tensor"]
        .detach()
        .cpu()
        .numpy()
    )

    require(
        actual_array.shape == (1, 3, 1024, 1024)
        and np.allclose(
            actual_array[0],
            expected_array,
            rtol=0,
            atol=1e-7,
        ),
        "Frozen RGB/255/CHW preprocessing parity failed.",
    )

    del first_source
    del first_crop
    del first_reload
    del prep_check
    del expected_array
    del actual_array

    state["preinference_png_and_preprocess_parity"] = (
        "PASS"
    )
    state["status"] = (
        "READY_FOR_FIRST_FROZEN_TILE_FORWARD"
    )

    atomic_json(state_path, state)

    print(
        "\nC. PASS lossless 1024 × 1024 tile bridge "
        "and frozen RGB preprocessing."
    )

    # ============================================================
    # D. EXECUTION DEVICE
    # ============================================================

    device = torch.device(
        "cuda:0"
        if torch.cuda.is_available()
        else "cpu"
    )

    print("\nDEVICE:", str(device))

    if device.type == "cuda":
        print(
            "CUDA GPU:",
            torch.cuda.get_device_name(device),
        )
    else:
        print("WARNING: CUDA unavailable.")

    state["runtime"] = {
        "pytorch": torch.__version__,
        "opencv": cv2.__version__,
        "numpy": np.__version__,
        "device": str(device),
        "cuda_available": torch.cuda.is_available(),
        "cuda_device_name": (
            torch.cuda.get_device_name(device)
            if device.type == "cuda"
            else None
        ),
    }

    atomic_json(state_path, state)

    # ============================================================
    # E. ALL THREE FROZEN SEEDS
    # ============================================================

    for (
        index,
        seed,
        threshold,
        filename,
        checkpoint_sha,
    ) in EXPECTED_SEEDS:

        key = f"seed{index:02d}_s{seed}"

        checkpoint_path = (
            BUNDLE
            / "final_checkpoints"
            / filename
        )

        verify(
            checkpoint_path,
            checkpoint_sha,
            f"Seed-{index} checkpoint",
        )

        print("\n" + "-" * 82)
        print(
            f"SEED {index}/3: seed={seed}, "
            f"operating threshold={threshold:.2f}"
        )

        print(
            "Checkpoint SHA256:",
            checkpoint_sha,
        )

        state["status"] = f"LOADING_{key}"
        atomic_json(state_path, state)

        # --------------------------------------------------------
        # RESUME03 FIX:
        #
        # The frozen RSC loader uses checkpoint_path.is_file().
        # It requires pathlib.Path, not str.
        # --------------------------------------------------------

        require(
            isinstance(checkpoint_path, Path),
            "Frozen checkpoint loader requires pathlib.Path.",
        )

        checkpoint = original.load_checkpoint_mapping(
            checkpoint_path
        )

        require(
            isinstance(checkpoint, dict),
            "Frozen checkpoint loader did not return a dict.",
        )

        model = phase5c.build_model(
            checkpoint,
            device,
        )

        require(
            isinstance(model, torch.nn.Module),
            "Frozen model builder did not return nn.Module.",
        )

        model = model.to(device)
        model.eval()

        del checkpoint

        seed_dir = run_dir / key

        seed_dir.mkdir(
            parents=True,
            exist_ok=False,
        )

        overlays = seed_dir / "overlays_all35"

        overlays.mkdir(
            parents=True,
            exist_ok=False,
        )

        floor_path = (
            seed_dir
            / "tile_decoder_floor_predictions.jsonl"
        )

        operating_path = (
            seed_dir
            / "image_operating_predictions.jsonl"
        )

        per_image_csv = (
            seed_dir
            / "qualitative_prediction_counts_per_image.csv"
        )

        seed_images_complete = 0
        seed_tile_count = 0
        seed_floor_count = 0
        seed_owned_floor_count = 0
        seed_operating_count = 0

        image_rows = []
        overlay_records = []

        predeclared = set(
            protocol["reported_outputs"][
                "predeclared_example_codes"
            ]
        )

        require(
            len(predeclared) == 6,
            "Predeclared six illustration samples changed.",
        )

        with (
            floor_path.open(
                "w",
                encoding="utf-8",
                buffering=1,
            ) as floor_handle,
            operating_path.open(
                "w",
                encoding="utf-8",
                buffering=1,
            ) as operating_handle,
        ):

            for code in sorted(source_by_code):
                source_info = source_by_code[code]

                full_image = cv2.imread(
                    str(source_info["image_path"]),
                    cv2.IMREAD_COLOR,
                )

                require(
                    full_image is not None
                    and full_image.shape == (2160, 3840, 3),
                    f"Unexpected source decode: {code}",
                )

                image_schedule = [
                    row
                    for row in schedule
                    if row["sample_code"] == code
                ]

                require(
                    len(image_schedule) == 12,
                    f"Expected 12 tiles for {code}.",
                )

                operating_predictions = []
                image_floor_count = 0
                image_owned_floor_count = 0

                for tile in image_schedule:
                    geometry = rect_from_schedule(tile)

                    x0 = geometry["tile_x0"]
                    y0 = geometry["tile_y0"]
                    x1 = geometry["tile_x1"]
                    y1 = geometry["tile_y1"]

                    tile_image = np.ascontiguousarray(
                        full_image[y0:y1, x0:x1, :]
                    )

                    require(
                        tile_image.shape == (1024, 1024, 3),
                        f"Non-1024 tile: {tile['tile_id']}",
                    )

                    require(
                        cv2.imwrite(
                            str(bridge_path),
                            tile_image,
                            [cv2.IMWRITE_PNG_COMPRESSION, 3],
                        ),
                        f"PNG bridge write failed: {tile['tile_id']}",
                    )

                    roundtrip = cv2.imread(
                        str(bridge_path),
                        cv2.IMREAD_COLOR,
                    )

                    require(
                        roundtrip is not None
                        and np.array_equal(
                            tile_image,
                            roundtrip,
                        ),
                        f"PNG bridge parity failed: {tile['tile_id']}",
                    )

                    # Original frozen RSC preprocessing, model
                    # forward, decoder, normalization and the
                    # separately frozen geometry-recovery policy.
                    predictions, prep = original.infer_one(
                        model=model,
                        decode_instances=decode_instances,
                        row={
                            "image_path": str(bridge_path),
                            "width": 1024,
                            "height": 1024,
                        },
                        device=device,
                        decoder_spec=FROZEN_DECODER,
                    )

                    require(
                        prep["original_w"] == 1024
                        and prep["original_h"] == 1024
                        and prep["padded_w"] == 1024
                        and prep["padded_h"] == 1024,
                        "Frozen preprocessing dimensions changed.",
                    )

                    require(
                        isinstance(predictions, list),
                        f"Unexpected decoder result: {tile['tile_id']}",
                    )

                    mapped = [
                        prediction_record(
                            prediction,
                            tile,
                            threshold,
                        )
                        for prediction in predictions
                    ]

                    owned = [
                        prediction
                        for prediction in mapped
                        if prediction[
                            "within_frozen_ownership_region"
                        ]
                    ]

                    kept = [
                        prediction
                        for prediction in owned
                        if prediction[
                            "passes_frozen_operating_threshold"
                        ]
                    ]

                    durable_jsonl_append(
                        floor_handle,
                        {
                            "sample_code": code,
                            "tile_id": tile["tile_id"],
                            "tile_x0": x0,
                            "tile_y0": y0,
                            "owner_region_xyxy": [
                                geometry["owner_x0"],
                                geometry["owner_y0"],
                                geometry["owner_x1"],
                                geometry["owner_y1"],
                            ],
                            "image_sha256": source_info[
                                "image_sha256"
                            ],
                            "checkpoint_sha256": checkpoint_sha,
                            "seed_index": index,
                            "decoder_floor": 0.05,
                            "frozen_operating_threshold": threshold,
                            "predictions_before_ownership_filter": mapped,
                            "owned_floor_prediction_count": len(owned),
                            "owned_operating_prediction_count": len(kept),
                        },
                    )

                    operating_predictions.extend(kept)

                    image_floor_count += len(mapped)
                    image_owned_floor_count += len(owned)

                    seed_tile_count += 1
                    seed_floor_count += len(mapped)
                    seed_owned_floor_count += len(owned)
                    seed_operating_count += len(kept)

                    state["completed_tile_forwards"] += 1

                    if seed_tile_count % 12 == 0:
                        state["status"] = f"RUNNING_{key}"
                        state["last_completed_tile_id"] = (
                            tile["tile_id"]
                        )
                        atomic_json(state_path, state)

                # No new cross-tile NMS or thresholding policy.
                # Only the pre-frozen ownership rule is applied.

                image_record = {
                    "sample_code": code,
                    "seed_index": index,
                    "seed": seed,
                    "checkpoint_sha256": checkpoint_sha,
                    "image_sha256": source_info[
                        "image_sha256"
                    ],
                    "original_size_wh": [3840, 2160],
                    "frozen_operating_threshold": threshold,
                    "tiles_evaluated": 12,
                    "tile_floor_prediction_count": image_floor_count,
                    "owned_floor_prediction_count": image_owned_floor_count,
                    "operating_owned_prediction_count": len(
                        operating_predictions
                    ),
                    "boxes": operating_predictions,
                    "interpretation": (
                        "QUALITATIVE_PREDICTIONS_NOT_VERIFIED_"
                        "PARTICLES_OR_PLASTIC_IDENTIFICATIONS"
                    ),
                }

                durable_jsonl_append(
                    operating_handle,
                    image_record,
                )

                overlay_path = (
                    overlays
                    / f"{code}__seed{index:02d}_qualitative.jpg"
                )

                overlay_sha = draw_overlay(
                    source_info["image_path"],
                    operating_predictions,
                    code,
                    index,
                    threshold,
                    overlay_path,
                )

                overlay_records.append({
                    "sample_code": code,
                    "path": str(overlay_path),
                    "sha256": overlay_sha,
                    "predeclared_illustration": (
                        code in predeclared
                    ),
                })

                image_rows.append({
                    "sample_code": code,
                    "seed_index": index,
                    "training_seed": seed,
                    "operating_threshold": threshold,
                    "tiles_evaluated": 12,
                    "tile_floor_count": image_floor_count,
                    "owned_floor_count": image_owned_floor_count,
                    "owned_operating_count": len(
                        operating_predictions
                    ),
                    "image_sha256": source_info[
                        "image_sha256"
                    ],
                    "overlay_sha256": overlay_sha,
                    "is_predeclared_illustration": (
                        code in predeclared
                    ),
                })

                seed_images_complete += 1
                state["completed_images"] += 1
                state["last_completed_sample_code"] = code

                atomic_json(state_path, state)

                print(
                    f"  {key}: "
                    f"{seed_images_complete:02d}/35 images, "
                    f"{seed_tile_count:03d}/420 tiles; "
                    f"{code} | "
                    f"owned operating boxes="
                    f"{len(operating_predictions)}",
                    flush=True,
                )

                del full_image

        require(
            seed_tile_count == 420
            and seed_images_complete == 35,
            f"Incomplete Seed-{index}: "
            f"{seed_tile_count} tiles, "
            f"{seed_images_complete} images.",
        )

        with per_image_csv.open(
            "w",
            encoding="utf-8",
            newline="",
        ) as handle:
            writer = csv.DictWriter(
                handle,
                fieldnames=list(image_rows[0].keys()),
                lineterminator="\n",
            )
            writer.writeheader()
            writer.writerows(image_rows)

        seed_summary = {
            "seed_index": index,
            "training_seed": seed,
            "frozen_operating_threshold": threshold,
            "checkpoint_sha256": checkpoint_sha,
            "images_complete": seed_images_complete,
            "tile_forwards": seed_tile_count,
            "tile_decoder_floor_box_count": seed_floor_count,
            "owned_decoder_floor_box_count": seed_owned_floor_count,
            "owned_frozen_operating_box_count": seed_operating_count,
            "decoder_floor_predictions_jsonl": {
                "path": str(floor_path),
                "sha256": sha256_file(floor_path),
            },
            "operating_predictions_jsonl": {
                "path": str(operating_path),
                "sha256": sha256_file(operating_path),
            },
            "per_image_csv": {
                "path": str(per_image_csv),
                "sha256": sha256_file(per_image_csv),
            },
            "overlays_all35": overlay_records,
            "reporting_scope": (
                "QUALITATIVE_EXTERNAL_MICROSCOPY_ONLY_"
                "NO_OBJECT_LEVEL_GT"
            ),
        }

        summary_path = (
            seed_dir
            / "TIOMAN_QUALITATIVE_SEED_SUMMARY_v1.json"
        )

        atomic_json(
            summary_path,
            seed_summary,
        )

        state["seed_results"][key] = {
            "summary": str(summary_path),
            "summary_sha256": sha256_file(
                summary_path
            ),
            "images": seed_images_complete,
            "tile_forwards": seed_tile_count,
            "owned_operating_box_count": seed_operating_count,
        }

        atomic_json(state_path, state)

        print(
            "SEED COMPLETE:",
            key,
            "| 420/420 tile forwards | summary SHA",
            sha256_file(summary_path),
            flush=True,
        )

        del model
        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    # ============================================================
    # F. FINAL COMPLETENESS VERIFICATION
    # ============================================================

    require(
        state["completed_tile_forwards"] == 1260,
        "All-seed tile forward count is not 1260.",
    )

    require(
        state["completed_images"] == 105,
        "All-seed image count is not 105.",
    )

    require(
        len(state["seed_results"]) == 3,
        "Three-seed result summaries incomplete.",
    )

    state["status"] = (
        "COMPLETE_1260_TILE_FORWARDS_"
        "ALL_THREE_FROZEN_SEEDS"
    )
    state["completed_utc"] = iso_now()
    state["scientific_scope"] = (
        "TIOMAN_35_IMAGES_UNANNOTATED_"
        "QUALITATIVE_EXTERNAL_MICROSCOPY_TRANSFER"
    )
    state["accuracy_metrics_computed"] = False
    state["note"] = (
        "Counts are model predictions, not object-level "
        "ground-truth correctness or verified polymer identities."
    )

    atomic_json(state_path, state)

    state_sha = sha256_file(state_path)

    sidecar = (
        run_dir
        / "TIOMAN_THREE_SEED_RUN_STATE_v1.sha256.txt"
    )

    sidecar.write_text(
        f"{state_sha}  {state_path.name}\n",
        encoding="utf-8",
    )

    print("\n" + "=" * 82)
    print(
        "TIOMAN INFERENCE STATUS:",
        state["status"],
    )
    print(
        "TILE FORWARDS:",
        state["completed_tile_forwards"],
    )
    print(
        "COMPLETE ORIGINAL IMAGE / SEED PAIRS:",
        state["completed_images"],
    )
    print("RUN DIRECTORY:", run_dir)
    print("STATE JSON:", state_path)
    print("STATE SHA256:", state_sha)

    for key, result in state["seed_results"].items():
        print(
            key,
            "summary_sha256=",
            result["summary_sha256"],
            "predicted_owned_operating_boxes=",
            result["owned_operating_box_count"],
        )

    print(
        "NO retraining, tuning, RSC re-evaluation, "
        "or GT-based accuracy metrics."
    )
    print(
        "Historical derived JSON byte discrepancies "
        "remain documented."
    )
    print("=" * 82)

except BaseException as exc:
    state["status"] = (
        "FAILED_OR_INTERRUPTED_OUTPUTS_PRESERVED"
    )
    state["failed_utc"] = iso_now()
    state["exception_type"] = type(exc).__name__
    state["exception_message"] = str(exc)
    state["exception_traceback"] = (
        traceback.format_exc()[-16000:]
    )

    atomic_json(state_path, state)

    failure_path = (
        run_dir
        / "TIOMAN_THREE_SEED_FAILURE_v1.json"
    )

    atomic_json(
        failure_path,
        {
            "status": state["status"],
            "run_directory": str(run_dir),
            "completed_tile_forwards": state[
                "completed_tile_forwards"
            ],
            "completed_images": state[
                "completed_images"
            ],
            "prior_seed_results": state[
                "seed_results"
            ],
            "error_type": state[
                "exception_type"
            ],
            "error": state[
                "exception_message"
            ],
            "traceback": state[
                "exception_traceback"
            ],
            "frozen_protocol_sha256": state[
                "frozen_protocol_sha256"
            ],
            "frozen_schedule_sha256": state[
                "frozen_schedule_sha256"
            ],
            "original_frozen_input_files_untouched": True,
            "prior_run_outputs_preserved": True,
            "resume_policy": (
                "Use a complete replacement script in a "
                "NEW resumeXX output directory. "
                "Never overwrite the failed run."
            ),
        },
    )

    print(
        "\nSTOP: Tioman inference failed or was interrupted. "
        "No frozen inputs modified.",
        flush=True,
    )
    print(
        "PRESERVED RUN DIR:",
        run_dir,
        flush=True,
    )
    print(
        "PRESERVED RUN STATE:",
        state_path,
        flush=True,
    )
    print(
        "PRESERVED FAILURE JSON:",
        failure_path,
        flush=True,
    )
    print(
        "FAILURE JSON SHA256:",
        sha256_file(failure_path),
        flush=True,
    )

    raise
```

```python

# ================================================================
# DASO-Net — TIOMAN AUDITED RESULTS PORTABLE FREEZE v1
# Complete standalone Kaggle cell. Evidence packaging ONLY.
# NO model loading, inference, tuning, training, or GT evaluation.
# Preserves all source run and audit files byte-for-byte.
# ================================================================

from pathlib import Path, PurePosixPath
from datetime import datetime, timezone
import csv
import hashlib
import json
import os
import traceback
import zipfile

WORK = Path("/kaggle/working")

RUN = WORK / "dasonet_tioman_frozen_three_seed_inference_v1_resume03"
AUDIT_DIR = WORK / "dasonet_tioman_results_integrity_audit_v1"
PROTOCOL_DIR = WORK / "dasonet_tioman_qualitative_tiling_protocol_freeze_v1"
PRIOR_DIR = WORK / "dasonet_tioman_preprotocol_interface_audit_v1_resume02"
INTERFACE_DIR = WORK / "dasonet_tioman_execution_interface_capture_v1"

CONT = Path(
    "/kaggle/input/datasets/abinavprabhuv/"
    "dasonet-continuation-checkpoint-v1/"
    "DASO_NET_CONTINUATION_CHECKPOINT_v1/"
    "dasonet_continuation_checkpoint_v1"
)

BUNDLE = CONT / "bundle"

RSC = (
    BUNDLE
    / "rsc_final/RSC_FORMAL_RECOVERY_RESULTS_FREEZE_v1/"
    "rsc_formal_recovery_results_freeze_v1/preserved_artifacts"
)

CODE = BUNDLE / "phase5b_code"

OUTPUT_BASE = WORK / "dasonet_tioman_audited_results_portable_freeze_v1"

EXPECTED = {
    "completed_run_state": (
        RUN / "TIOMAN_THREE_SEED_RUN_STATE_v1.json",
        "d6ffb0c2ecdd418872ea4f5e44fe0c65856a10e74c0b348b7d5c1a5ac5315a05",
    ),
    "integrity_audit": (
        AUDIT_DIR / "TIOMAN_THREE_SEED_RESULTS_INTEGRITY_AUDIT_v1.json",
        "580c41a77312a616ec94b5b7e32a61e284a47c925e30a16595e7f882a685247f",
    ),
    "tioman_protocol": (
        PROTOCOL_DIR / "TIOMAN_QUALITATIVE_TILING_PROTOCOL_FREEZE_v1.json",
        "75b81b88550d7f76f6e7cecf27c4e805a281e350a99159ebba8c4447daaae563",
    ),
    "tile_schedule": (
        PROTOCOL_DIR / "TIOMAN_TILE_SCHEDULE_v1.csv",
        "4aa24b8864052108146d4df9e1db490569bed1172f8b12e78aa95260b1c70f7f",
    ),
    "resume02_source_audit": (
        PRIOR_DIR / "TIOMAN_PREPROTOCOL_INTERFACE_AUDIT_RESUME02.json",
        "ca505ee2597dcf75adb575ec81e64d7918fb4ec2edf1f24583ab96ea0a938e91",
    ),
    "execution_interface": (
        INTERFACE_DIR / "TIOMAN_FROZEN_EXECUTION_INTERFACE_CAPTURE_v1.json",
        "20ad2da7a61fed515ea9f565791a20b4931c92859d59e0917c5473dc9e3403a0",
    ),
    "rsc_original_evaluator": (
        RSC / "05_original_evaluator_implementation_freeze/evaluate_rsc_external_v1_FROZEN.py",
        "59a3de31f2a5244ba32c53b705b050e033b5d17c8de7efb36c17dbc49f0a7723",
    ),
    "rsc_recovery_evaluator": (
        RSC / "12_recovery_evaluator_implementation_freeze/evaluate_rsc_external_recovery_v1_FROZEN.py",
        "d0df4ad7cdd60aae23fdff7d2c708c938952510df4a8d9d7a9087eecbad85bda",
    ),
    "phase5c_evaluator": (
        CODE / "scripts/evaluate_phase5c.py",
        "9c6221b081121f34c0008c1ff49205758d5e551c01641b52de7643e5cf7e4e14",
    ),
    "model_definition": (
        CODE / "src/dasonet/model/dasonet.py",
        "1ac6f92e1683c6939ff5783ac0cae065dab20e366dd49a9b72ac89007e5eaa40",
    ),
    "decoder": (
        CODE / "src/dasonet/model/decoding.py",
        "9799eb1be568e3d6e0ce82b7c99df7a0ed05d271cd44ff24b612526cc18537d1",
    ),
    "rsc_protocol": (
        RSC / "02_protocol_freeze/RSC_EXTERNAL_EVALUATION_PROTOCOL_FREEZE_v1.json",
        "e3c87caebafe87c7671b5745a480c0edf67f444feac4278633da6addae4d1f7e",
    ),
    "rsc_recovery_policy": (
        RSC / "11_recovery_policy_freeze/RSC_DEGENERATE_BOX_RECOVERY_POLICY_FREEZE_v1.json",
        "aa754df77f897c68421a065ffa0b0ffcfe6b332b8dba96b6415d089cb2216f8b",
    ),
    "source_manifest": (
        BUNDLE / "frozen_manifests/public_real_mp_v1_source_specific_manifest.csv",
        "1bc39ae0ecb03436b7948138a6a7f28547d0d610e116442328d7e5dad0a46301",
    ),
}

SEEDS = [
    (
        1, 1782656963, 0.30, 499, 34,
        "6b7e3191c1363cbcb99ecaefe7df35d979ad48c39457ce2891f35cc8c7b04923",
    ),
    (
        2, 901395089, 0.35, 58, 17,
        "4dcda8b83b2970b86a9504b468e4220ad5c3d8f7536782f84e96293c184bdd26",
    ),
    (
        3, 252916394, 0.35, 135, 25,
        "d0b0974255d900979931b784ecb1e1b74a0d3868beb0952126ffa510752be52f",
    ),
]


def need(condition, message):
    if not condition:
        raise RuntimeError(message)


def hash_file(path):
    digest = hashlib.sha256()
    with Path(path).open("rb") as handle:
        for chunk in iter(
            lambda: handle.read(4 * 1024 * 1024), b""
        ):
            digest.update(chunk)
    return digest.hexdigest()


def verified(path, expected, label):
    need(path.is_file(), f"Missing {label}: {path}")
    actual = hash_file(path)
    need(
        actual == expected,
        f"SHA-256 mismatch for {label}: "
        f"{actual} != {expected}",
    )
    return actual


def read_json(path):
    return json.loads(
        path.read_text(encoding="utf-8")
    )


def encoded_json(obj):
    return (
        json.dumps(
            obj,
            indent=2,
            sort_keys=True,
            ensure_ascii=False,
            allow_nan=False,
        ) + "\n"
    ).encode("utf-8")


def fresh_dir(base):
    for number in range(100):
        candidate = (
            base
            if number == 0
            else Path(str(base) + f"_resume{number:02d}")
        )
        if not candidate.exists():
            candidate.mkdir(parents=True, exist_ok=False)
            return candidate
    raise RuntimeError(
        "No unused freeze directory; preserve previous attempts."
    )


def validate_member(member):
    path = PurePosixPath(member)
    need(
        not path.is_absolute()
        and all(part not in ("", ".", "..") for part in path.parts),
        "Unsafe ZIP member: " + member,
    )
    need(
        path.as_posix() == member,
        "Noncanonical ZIP member: " + member,
    )


def fixed_zipinfo(member):
    info = zipfile.ZipInfo(
        member, date_time=(1980, 1, 1, 0, 0, 0)
    )
    info.compress_type = zipfile.ZIP_DEFLATED
    info.external_attr = 0o100644 << 16
    info.create_system = 3
    return info


print("=" * 78)
print("DASO-Net — TIOMAN AUDITED RESULTS PORTABLE FREEZE v1")
print("READ-ONLY SCIENTIFIC INPUTS | NO INFERENCE")
print("=" * 78)

# ================================================================
# A. VERIFY FROZEN IDENTITIES
# ================================================================

for key, (path, digest) in EXPECTED.items():
    verified(path, digest, key)

print("PASS 14 pinned provenance, protocol, and code identities")

state = read_json(EXPECTED["completed_run_state"][0])
audit = read_json(EXPECTED["integrity_audit"][0])
protocol = read_json(EXPECTED["tioman_protocol"][0])

need(
    state["status"]
    == "COMPLETE_1260_TILE_FORWARDS_ALL_THREE_FROZEN_SEEDS",
    "Completed inference state is not PASS.",
)

need(
    state["completed_tile_forwards"] == 1260
    and state["completed_images"] == 105,
    "Completed inference totals changed.",
)

need(
    audit["status"]
    == "PASS_ALL_THREE_SEED_OUTPUTS_INDEPENDENTLY_RECONCILED",
    "Independent results-integrity audit is not PASS.",
)

need(
    audit["source_run_file_count"] == 120
    and audit["combined_seed_prediction_records"] == 692,
    "Independent audit totals changed.",
)

need(
    audit["historical_json_byte_discrepancies_preserved"] is True,
    "Historical audit discrepancy record absent.",
)

need(
    len(protocol["images"]) == 35
    and len(protocol["final_seeds"]) == 3,
    "Frozen Tioman image/seed population changed.",
)

# ================================================================
# B. VERIFY INDEPENDENT AUDIT FILES
# ================================================================

AUDIT_COMPONENTS = {
    "source_manifest": (
        AUDIT_DIR / "TIOMAN_SOURCE_RUN_FILE_HASH_MANIFEST_v1.json"
    ),
    "counts_csv": (
        AUDIT_DIR / "TIOMAN_THREE_SEED_PER_IMAGE_AUDITED_COUNTS_v1.csv"
    ),
    "audit_sha_sidecar": (
        AUDIT_DIR / "TIOMAN_THREE_SEED_RESULTS_INTEGRITY_AUDIT_v1.sha256.txt"
    ),
}

verified(
    AUDIT_COMPONENTS["source_manifest"],
    audit["derived_outputs"]["source_run_file_manifest_sha256"],
    "independent source-run file manifest",
)

verified(
    AUDIT_COMPONENTS["counts_csv"],
    audit["derived_outputs"]["audited_per_image_csv_sha256"],
    "audited image counts CSV",
)

need(
    AUDIT_COMPONENTS["audit_sha_sidecar"]
    .read_text(encoding="utf-8")
    .strip()
    .split()[0]
    == EXPECTED["integrity_audit"][1],
    "Integrity audit SHA sidecar mismatch.",
)

with AUDIT_COMPONENTS["counts_csv"].open(
    "r", encoding="utf-8", newline=""
) as handle:
    count_rows = list(csv.DictReader(handle))

need(
    len(count_rows) == 105,
    "Audited image/seed records != 105.",
)

for idx, seed, threshold, boxes, positives, summary_sha in SEEDS:
    key = f"seed{idx:02d}_s{seed}"

    summary = (
        RUN / key / "TIOMAN_QUALITATIVE_SEED_SUMMARY_v1.json"
    )
    verified(summary, summary_sha, key + " summary")

    need(
        state["seed_results"][key]["summary_sha256"] == summary_sha,
        key + " summary identity mismatch.",
    )

    relevant = [
        row for row in count_rows
        if int(row["seed_index"]) == idx
    ]

    need(len(relevant) == 35, key + " image count mismatch.")

    need(
        sum(
            int(row["owned_operating_boxes"])
            for row in relevant
        ) == boxes,
        key + " prediction total mismatch.",
    )

    need(
        sum(
            int(row["owned_operating_boxes"]) > 0
            for row in relevant
        ) == positives,
        key + " positive-image total mismatch.",
    )

    need(
        all(
            float(row["operating_threshold"]) == threshold
            for row in relevant
        ),
        key + " threshold mismatch.",
    )

    need(
        state["seed_results"][key]["owned_operating_box_count"]
        == boxes,
        key + " completed-state count mismatch.",
    )

print("PASS 105 audited image/seed rows and 3 seed summaries")

# ================================================================
# C. REVERIFY ALL 120 SOURCE RUN FILES
# ================================================================

source_manifest = read_json(
    AUDIT_COMPONENTS["source_manifest"]
)

need(
    source_manifest["source_run"] == str(RUN),
    "Manifest references another run.",
)

files = source_manifest["files"]

need(
    isinstance(files, list) and len(files) == 120,
    "Source-run manifest does not contain 120 files.",
)

listed = [item["relative_path"] for item in files]

need(
    len(set(listed)) == 120,
    "Duplicate source-run relative path.",
)

actual = {
    path.relative_to(RUN).as_posix()
    for path in RUN.rglob("*")
    if path.is_file()
}

need(
    actual == set(listed),
    "Source run contains missing or unmanifested files.",
)

payload = {}


def register(member, source_path, expected_sha=None, expected_size=None):
    validate_member(member)

    need(
        member not in payload,
        "Duplicate archive member: " + member,
    )

    source_path = Path(source_path)
    need(
        source_path.is_file(),
        "Missing source artifact: " + str(source_path),
    )

    size = source_path.stat().st_size

    if expected_size is not None:
        need(
            size == int(expected_size),
            "Source size mismatch: " + member,
        )

    digest = hash_file(source_path)

    if expected_sha is not None:
        need(
            digest == expected_sha,
            "Source SHA mismatch: " + member,
        )

    payload[member] = {
        "source_path": source_path,
        "sha256": digest,
        "size_bytes": size,
    }


for item in files:
    relative = item["relative_path"]
    validate_member(relative)

    source = RUN / relative

    need(
        source.resolve().is_relative_to(RUN.resolve()),
        "Source-run file escapes directory: " + relative,
    )

    register(
        "run/" + relative,
        source,
        item["sha256"],
        item["size_bytes"],
    )

print("PASS 120/120 run artifacts independently rehashed")

# ================================================================
# D. ADD SCIENTIFIC AUDIT AND REFERENCE RECORDS
# ================================================================

for name in (
    "TIOMAN_THREE_SEED_RESULTS_INTEGRITY_AUDIT_v1.json",
    "TIOMAN_SOURCE_RUN_FILE_HASH_MANIFEST_v1.json",
    "TIOMAN_THREE_SEED_PER_IMAGE_AUDITED_COUNTS_v1.csv",
    "TIOMAN_THREE_SEED_RESULTS_INTEGRITY_AUDIT_v1.sha256.txt",
):
    register(
        "audit/" + name,
        AUDIT_DIR / name,
    )

for label in ("tioman_protocol", "tile_schedule"):
    path, digest = EXPECTED[label]
    register("protocol/" + path.name, path, digest)

for label in ("resume02_source_audit", "execution_interface"):
    path, digest = EXPECTED[label]
    register("provenance/" + path.name, path, digest)

for label in (
    "rsc_original_evaluator",
    "rsc_recovery_evaluator",
    "phase5c_evaluator",
    "model_definition",
    "decoder",
    "rsc_protocol",
    "rsc_recovery_policy",
    "source_manifest",
):
    path, digest = EXPECTED[label]
    register(
        "frozen_reference/" + label + "__" + path.name,
        path,
        digest,
    )

# Verify that the six predeclared figure samples are present
# for all three seeds before packaging.

six_codes = protocol["reported_outputs"][
    "predeclared_example_codes"
]

need(
    len(six_codes) == len(set(six_codes)) == 6,
    "Six predeclared examples changed.",
)

figure_members = [
    (
        f"run/seed{idx:02d}_s{seed}/overlays_all35/"
        f"{code}__seed{idx:02d}_qualitative.jpg"
    )
    for idx, seed, *_ in SEEDS
    for code in six_codes
]

need(
    len(figure_members) == 18
    and all(member in payload for member in figure_members),
    "Not all 18 predeclared figure overlays are present.",
)

print("PASS 18/18 predeclared figure overlays")

# ================================================================
# E. SCIENTIFIC SCOPE AND PACKAGE FREEZE RECORD
# ================================================================

readme = f"""DASO-Net | Tioman audited qualitative results | evidence package v1

STATUS: Independent results-integrity audit PASS.
All 1260 frozen tile forwards completed using three frozen checkpoints
over 35 original images.

Frozen operating-threshold predictions:
Seed 1 = 499; Seed 2 = 58; Seed 3 = 135.

Total = 692 SEED-SPECIFIC prediction records,
NOT 692 distinct verified particles.

Tioman protocol SHA256: {EXPECTED['tioman_protocol'][1]}
Tile schedule SHA256: {EXPECTED['tile_schedule'][1]}
Inference state SHA256: {EXPECTED['completed_run_state'][1]}
Integrity audit SHA256: {EXPECTED['integrity_audit'][1]}

Archive directories:

run/               All 120 completed-run files, including 105 overlays.
audit/             Independent audit, audited counts, hash manifest.
protocol/          Pre-inference frozen protocol and tile schedule.
provenance/        Previous evidence and execution-interface audits.
frozen_reference/  Immutable reference evaluator and decoder sources.

Predeclared six qualitative illustration samples:
{', '.join(six_codes)}

Their 18 original seed overlays are included under run/.

SCIENTIFIC LIMITATIONS

Microscopy images lack established particle-level boxes/masks.
RGB predictions do not establish polymer or particle identity.
FTIR candidate reports are sample-linked, not per-object GT.
No Tioman detection AP, precision, recall, F1, Dice, or chemical
classification accuracy is established.

Between-seed disagreement is descriptive, not seed selection.
Original JPEG decoding warnings remain documented.
Two historical derived JSON SHA discrepancies remain documented
in Resume02; their causes have not been established.

Earlier failed inference runs remain preserved separately in Kaggle.
This archive packages successful Resume03 and its verified audit.

NOT INCLUDED

Original Tioman JPG/PDF/ISPD/XLSX source dataset.
Three original checkpoint PT binaries.
The executed Resume03 notebook cell.

These remain external dependencies for complete reproduction.

The package content manifest hashes every payload entry except
the manifest itself, avoiding circular self-reference.

No training, model loading, tuning, or inference was performed
to create this evidence archive.
"""

freeze_record = {
    "freeze_name": (
        "TIOMAN_AUDITED_QUALITATIVE_RESULTS_PORTABLE_FREEZE_v1"
    ),
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "status": "VERIFIED_SOURCE_OUTPUTS_READY_TO_PACKAGE",
    "completed_run_state_sha256": EXPECTED["completed_run_state"][1],
    "independent_results_integrity_audit_sha256": (
        EXPECTED["integrity_audit"][1]
    ),
    "frozen_tioman_protocol_sha256": EXPECTED["tioman_protocol"][1],
    "frozen_tile_schedule_sha256": EXPECTED["tile_schedule"][1],
    "source_run_file_count": 120,
    "original_images": 35,
    "image_seed_pairs": 105,
    "tile_forwards": 1260,
    "seed_specific_prediction_records": 692,
    "seed_breakdown": [
        {
            "seed_index": idx,
            "training_seed": seed,
            "operating_threshold": threshold,
            "predicted_boxes": boxes,
            "images_with_at_least_one_prediction": positives,
            "seed_summary_sha256": summary_sha,
        }
        for idx, seed, threshold, boxes, positives, summary_sha
        in SEEDS
    ],
    "predeclared_figure_samples": six_codes,
    "predeclared_figure_overlay_count": 18,
    "historical_derived_audit_json_byte_discrepancies_preserved": True,
    "raw_microscopy_or_ftir_data_included": False,
    "checkpoint_binaries_included": False,
    "executed_notebook_cell_included": False,
    "object_level_ground_truth": False,
    "scientific_evaluation": (
        "QUALITATIVE_RGB_MICROSCOPY_TRANSFER_ONLY"
    ),
    "model_loaded": False,
    "inference_performed": False,
    "training_performed": False,
    "threshold_tuning_performed": False,
}

# ================================================================
# F. WRITE FRESH ZIP WITHOUT OVERWRITING PREVIOUS ATTEMPTS
# ================================================================

OUT = fresh_dir(OUTPUT_BASE)
print("NEW FREEZE DIRECTORY:", OUT)

try:
    readme_bytes = readme.encode("utf-8")
    freeze_bytes = encoded_json(freeze_record)

    memory_payload = {
        "README_SCIENTIFIC_SCOPE_v1.txt": readme_bytes,
        "TIOMAN_AUDITED_RESULTS_FREEZE_v1.json": freeze_bytes,
    }

    for member in memory_payload:
        need(
            member not in payload,
            "Metadata path conflicts with source payload.",
        )

    full_inventory = [
        {
            "archive_path": member,
            "size_bytes": item["size_bytes"],
            "sha256": item["sha256"],
        }
        for member, item in sorted(payload.items())
    ]

    full_inventory += [
        {
            "archive_path": member,
            "size_bytes": len(content),
            "sha256": hashlib.sha256(content).hexdigest(),
        }
        for member, content in sorted(memory_payload.items())
    ]

    full_inventory.sort(
        key=lambda item: item["archive_path"]
    )

    manifest_bytes = encoded_json({
        "manifest_name": (
            "TIOMAN_AUDITED_RESULTS_PACKAGE_CONTENT_MANIFEST_v1"
        ),
        "manifest_itself_excluded_from_digest_records": True,
        "payload_entry_count": len(full_inventory),
        "files": full_inventory,
    })

    manifest_name = (
        "TIOMAN_AUDITED_RESULTS_PACKAGE_CONTENT_MANIFEST_v1.json"
    )

    memory_payload[manifest_name] = manifest_bytes

    zip_path = (
        OUT / "DASO_Net_Tioman_Audited_Results_Evidence_v1.zip"
    )

    partial_zip = OUT / (zip_path.name + ".incomplete")

    with zipfile.ZipFile(
        partial_zip, "w", allowZip64=True
    ) as archive:

        for member in sorted(
            set(payload) | set(memory_payload)
        ):
            if member in payload:
                info = fixed_zipinfo(member)

                with (
                    payload[member]["source_path"].open("rb") as source,
                    archive.open(
                        info, "w", force_zip64=True
                    ) as destination,
                ):
                    for chunk in iter(
                        lambda: source.read(4 * 1024 * 1024),
                        b"",
                    ):
                        destination.write(chunk)

            else:
                archive.writestr(
                    fixed_zipinfo(member),
                    memory_payload[member],
                )

    # Verify ZIP CRC, member population, and each member's SHA256.

    with zipfile.ZipFile(partial_zip, "r") as archive:
        archive_names = archive.namelist()

        expected_names = (
            set(payload) | set(memory_payload)
        )

        need(
            len(archive_names) == len(expected_names)
            and set(archive_names) == expected_names,
            "ZIP member population mismatch.",
        )

        bad_crc_member = archive.testzip()

        need(
            bad_crc_member is None,
            "ZIP CRC failure: " + str(bad_crc_member),
        )

        for item in full_inventory:
            with archive.open(
                item["archive_path"], "r"
            ) as source:

                digest = hashlib.sha256()
                byte_count = 0

                for chunk in iter(
                    lambda: source.read(4 * 1024 * 1024),
                    b"",
                ):
                    digest.update(chunk)
                    byte_count += len(chunk)

            need(
                digest.hexdigest() == item["sha256"]
                and byte_count == item["size_bytes"],
                "Archived member verification failed: "
                + item["archive_path"],
            )

        need(
            archive.read(manifest_name) == manifest_bytes,
            "Embedded content manifest mismatch.",
        )

    os.replace(partial_zip, zip_path)

    zip_sha = hash_file(zip_path)

    zip_sidecar = OUT / (zip_path.name + ".sha256.txt")

    zip_sidecar.write_text(
        f"{zip_sha}  {zip_path.name}\n",
        encoding="utf-8",
    )

    verification = {
        "status": (
            "PORTABLE_RESULTS_FREEZE_COMPLETE_"
            "ZIP_READBACK_VERIFIED"
        ),
        "zip_name": zip_path.name,
        "zip_size_bytes": zip_path.stat().st_size,
        "zip_sha256": zip_sha,
        "zip_crc_test": "PASS",
        "zip_file_member_sha256_readback": "PASS",
        "package_payload_entries_excluding_manifest": (
            len(full_inventory)
        ),
        "archive_members_including_manifest": (
            len(expected_names)
        ),
        "source_run_files": 120,
        "predeclared_qualitative_overlays": 18,
        "training_or_inference_run": False,
        "checkpoint_binaries_included": False,
        "raw_tioman_source_assets_included": False,
    }

    verification_path = (
        OUT / "TIOMAN_AUDITED_RESULTS_PACKAGE_VERIFICATION_v1.json"
    )

    verification_path.write_bytes(
        encoded_json(verification)
    )

    print("=" * 78)
    print(
        "PORTABLE FREEZE STATUS:",
        verification["status"],
    )
    print("INFERENCE / TRAINING / TUNING: NO")
    print(
        "SOURCE RUN FILES VERIFIED AND PACKAGED:",
        verification["source_run_files"],
    )
    print(
        "TOTAL ZIP MEMBERS:",
        verification["archive_members_including_manifest"],
    )
    print(
        "PREDECLARED FIGURE OVERLAYS INCLUDED:",
        verification["predeclared_qualitative_overlays"],
    )
    print("ZIP:", zip_path)
    print("ZIP SIZE BYTES:", zip_path.stat().st_size)
    print("ZIP SHA256:", zip_sha)
    print("ZIP SHA RECORD:", zip_sidecar)
    print("VERIFICATION JSON:", verification_path)
    print(
        "Save the ZIP and SHA sidecar to persistent storage; "
        "/kaggle/working is temporary."
    )
    print("=" * 78)

except BaseException as exc:
    failure_path = (
        OUT / "TIOMAN_AUDITED_RESULTS_PACKAGE_FAILURE_v1.json"
    )

    failure_path.write_bytes(
        encoded_json({
            "status": "FAILED_PACKAGE_ATTEMPT_PRESERVED",
            "error_type": type(exc).__name__,
            "error": str(exc),
            "traceback": traceback.format_exc()[-15000:],
            "source_run_unchanged": True,
            "previous_attempts_preserved": True,
        })
    )

    print(
        "PACKAGE FAILED. PRESERVED FAILURE:",
        failure_path,
    )
    raise
```

```python
# ================================================================
# DASO-Net — TIOMAN NATIVE-RESOLUTION VISUAL AUDIT v1
# Complete standalone Kaggle notebook cell, read-only frozen inputs.
# Creates exact 384x384 native-pixel crop pairs for EVERY prediction
# in the six predeclared images, plus paginated review contact sheets.
# No inference, threshold tuning, human labels, or GT metrics.
# ================================================================
from pathlib import Path
from datetime import datetime, timezone
from collections import defaultdict
import csv, hashlib, json, math, os, traceback, zipfile
import cv2
import numpy as np

WORK = Path('/kaggle/working')
RUN = WORK/'dasonet_tioman_frozen_three_seed_inference_v1_resume03'
FREEZE = WORK/'dasonet_tioman_qualitative_tiling_protocol_freeze_v1'
AUDIT = WORK/'dasonet_tioman_results_integrity_audit_v1'
PORTABLE = WORK/'dasonet_tioman_audited_results_portable_freeze_v1'
REVIEW = WORK/'dasonet_tioman_predeclared_visual_review_v1'
OUTBASE = WORK/'dasonet_tioman_native_resolution_visual_audit_v1'

PIN = {
 'protocol':(FREEZE/'TIOMAN_QUALITATIVE_TILING_PROTOCOL_FREEZE_v1.json','75b81b88550d7f76f6e7cecf27c4e805a281e350a99159ebba8c4447daaae563'),
 'schedule':(FREEZE/'TIOMAN_TILE_SCHEDULE_v1.csv','4aa24b8864052108146d4df9e1db490569bed1172f8b12e78aa95260b1c70f7f'),
 'run':(RUN/'TIOMAN_THREE_SEED_RUN_STATE_v1.json','d6ffb0c2ecdd418872ea4f5e44fe0c65856a10e74c0b348b7d5c1a5ac5315a05'),
 'audit':(AUDIT/'TIOMAN_THREE_SEED_RESULTS_INTEGRITY_AUDIT_v1.json','580c41a77312a616ec94b5b7e32a61e284a47c925e30a16595e7f882a685247f'),
 'portable':(PORTABLE/'DASO_Net_Tioman_Audited_Results_Evidence_v1.zip','2cad35c34469d5d59863b7f8456535003ba8062c426c7cdeecd8ffd5ca25cfc2'),
 'visual_prep':(REVIEW/'TIOMAN_PREDECLARED_VISUAL_REVIEW_PREPARATION_v1.json','c1435b5bfd9fdf34fa7632e8bf6e770d0b023df80102a244f716c124095bb8ff'),
}
SEEDS = (
 (1,1782656963,0.30,499,'6b7e3191c1363cbcb99ecaefe7df35d979ad48c39457ce2891f35cc8c7b04923'),
 (2,901395089,0.35,58,'4dcda8b83b2970b86a9504b468e4220ad5c3d8f7536782f84e96293c184bdd26'),
 (3,252916394,0.35,135,'d0b0974255d900979931b784ecb1e1b74a0d3868beb0952126ffa510752be52f'),
)
SIX = ('JB_S01_B_F_1','JB_S02_A_M_1','SLB_S01_A_M_1','SLB_S02_A_F_1','TB_S01_A_F_1','TB_S01_A_F_2')
SIX_COUNTS = {
 'JB_S01_B_F_1':(30,0,0), 'JB_S02_A_M_1':(6,1,2),
 'SLB_S01_A_M_1':(2,3,10), 'SLB_S02_A_F_1':(0,1,1),
 'TB_S01_A_F_1':(5,3,6), 'TB_S01_A_F_2':(2,1,1),
}
CROP = 384
PER_PAGE = 8
PAGE_COLS = 2
CARD_W = CROP*2+20
CARD_H = CROP+54


def need(condition, message):
    if not condition:
        raise RuntimeError(message)


def sha(path):
    h=hashlib.sha256()
    with Path(path).open('rb') as f:
        for chunk in iter(lambda:f.read(4*1024*1024),b''):
            h.update(chunk)
    return h.hexdigest()


def checked(path, digest, label):
    need(Path(path).is_file(),f'Missing {label}: {path}')
    actual=sha(path)
    need(actual==digest,f'{label} SHA mismatch: {actual} != {digest}')


def getj(path):
    return json.loads(Path(path).read_text(encoding='utf-8'))


def fresh(base):
    for n in range(100):
        path=base if n==0 else Path(str(base)+f'_resume{n:02d}')
        if not path.exists():
            path.mkdir(parents=True,exist_ok=False)
            return path
    raise RuntimeError('No free output directory; prior results preserved')


def dumpj(path, data):
    Path(path).write_text(json.dumps(data,indent=2,sort_keys=True,ensure_ascii=False,allow_nan=False)+'\n',encoding='utf-8')


def png_write(path, image):
    ok=cv2.imwrite(str(path),image,[cv2.IMWRITE_PNG_COMPRESSION,3])
    need(ok,f'Failed to write PNG: {path}')


def roi_start(cx, cy):
    # No resizing: exactly 384 original JPEG-decoded pixels per axis.
    left=max(0,min(3840-CROP,int(math.floor(cx-CROP/2))))
    top=max(0,min(2160-CROP,int(math.floor(cy-CROP/2))))
    return left, top


def make_card(source, p, code, seed_index, seq):
    b=list(map(float,p['image_box_xyxy']))
    need(len(b)==4 and all(map(math.isfinite,b)), 'Invalid predicted coordinates')
    need(0<=b[0]<b[2]<=3840 and 0<=b[1]<b[3]<=2160,'Prediction out of image bounds')
    sc=float(p['score'])
    need(math.isfinite(sc) and 0<=sc<=1,'Invalid prediction score')
    cx,cy=(b[0]+b[2])/2,(b[1]+b[3])/2
    need(all(abs(float(a)-c)<1e-5 for a,c in zip(p['original_center_xy'],(cx,cy))),'Original-center provenance changed')
    x,y=roi_start(cx,cy)
    original=np.ascontiguousarray(source[y:y+CROP,x:x+CROP,:])
    need(original.shape==(CROP,CROP,3),'Native crop dimension failure')
    annotated=original.copy()
    p1=(max(0,min(CROP-1,int(round(b[0]-x)))), max(0,min(CROP-1,int(round(b[1]-y)))))
    p2=(max(0,min(CROP-1,int(round(b[2]-x)))), max(0,min(CROP-1,int(round(b[3]-y)))))
    cv2.rectangle(annotated,p1,p2,(0,255,255),2)
    center=(max(0,min(CROP-1,int(round(cx-x)))),max(0,min(CROP-1,int(round(cy-y)))))
    cv2.circle(annotated,center,6,(255,0,255),1)
    card=np.zeros((CARD_H,CARD_W,3),dtype=np.uint8)
    card[50:50+CROP,5:5+CROP,:]=original
    card[50:50+CROP,15+CROP:15+2*CROP,:]=annotated
    caption=f'{code} | seed{seed_index} box{seq:03d} | score {sc:.4f} | native pixels'
    cv2.putText(card,caption[:102],(8,22),cv2.FONT_HERSHEY_SIMPLEX,0.45,(255,255,255),1,cv2.LINE_AA)
    cv2.putText(card,'ORIGINAL CROP', (10,42),cv2.FONT_HERSHEY_SIMPLEX,0.43,(255,255,255),1)
    cv2.putText(card,'SAME CROP + BOX (RENDERED)',(CROP+20,42),cv2.FONT_HERSHEY_SIMPLEX,0.43,(255,255,255),1)
    metadata={'sample_code':code,'seed_index':seed_index,'box_index_in_image_seed':seq,
              'score':sc,'image_box_xyxy':b,'crop_xyxy':[x,y,x+CROP,y+CROP],
              'tile_id':p['tile_id'],'decoder_order':int(p['decoder_order']),
              'native_pixels_unscaled':True,'rendered_box_not_ground_truth':True}
    return card, metadata


OUT=fresh(OUTBASE)
print('='*78)
print('DASO-Net — NATIVE-RESOLUTION SIX-SAMPLE REVIEW AUDIT v1')
print('NEW DERIVED OUTPUT:',OUT)
print('READ-ONLY frozen results; NO inference/labels/tuning')
print('='*78)

try:
    for label,(path,digest) in PIN.items():
        checked(path,digest,label)
        print('PASS',label,digest[:16])

    protocol=getj(PIN['protocol'][0]); state=getj(PIN['run'][0]); audit=getj(PIN['audit'][0]); prep=getj(PIN['visual_prep'][0])
    need(state['status']=='COMPLETE_1260_TILE_FORWARDS_ALL_THREE_FROZEN_SEEDS','Inference run not completed')
    need(audit['status']=='PASS_ALL_THREE_SEED_OUTPUTS_INDEPENDENTLY_RECONCILED','Independent results audit not passed')
    need(tuple(protocol['reported_outputs']['predeclared_example_codes'])==SIX,'Predeclared six samples changed')
    need(tuple(prep['frozen_predeclared_sample_codes_in_order'])==SIX,'Preview selection provenance differs')
    need(prep['source_package_sha256']==PIN['portable'][1],'Preview references different archive')

    images={item['sample_code']:item for item in protocol['images']}
    need(len(images)==35 and set(SIX).issubset(images),'Frozen original image population differs')

    previews={item['sample_code']:item for item in prep['figures']}
    need(len(previews)==6,'Predeclared comparison figure count differs')
    comparison_dir=OUT/'verified_four_panel_previews'; comparison_dir.mkdir()
    for code in SIX:
        figure=previews[code]
        original=REVIEW/figure['filename']
        checked(original,figure['sha256'],f'{code} montage')
        (comparison_dir/original.name).write_bytes(original.read_bytes())

    by_seed={}
    for index,trainseed,threshold,total,summarysha in SEEDS:
        key=f'seed{index:02d}_s{trainseed}'
        s_path=RUN/key/'TIOMAN_QUALITATIVE_SEED_SUMMARY_v1.json'
        checked(s_path,summarysha,key+' summary')
        s=getj(s_path)
        need(s['tile_forwards']==420 and s['images_complete']==35 and s['owned_frozen_operating_box_count']==total, key+' counts mismatch')
        need(float(s['frozen_operating_threshold'])==threshold, key+' threshold differs')
        pred_path=RUN/key/'image_operating_predictions.jsonl'
        checked(pred_path,s['operating_predictions_jsonl']['sha256'],key+' predictions')
        with pred_path.open('r',encoding='utf-8') as f:
            rows=[json.loads(line) for line in f if line.strip()]
        need(len(rows)==35 and len({r['sample_code'] for r in rows})==35, key+' image prediction rows incomplete')
        by_seed[index]={r['sample_code']:r for r in rows}
        need(sum(len(r['boxes']) for r in rows)==total,key+' aggregate operating counts differ')

    out_pages=OUT/'native_crop_atlas_pages'; out_pages.mkdir()
    all_meta=[]; atlas_manifest=[]; expected_total=0
    for code in SIX:
        im=images[code]
        path=Path(im['image_path'])
        checked(path,im['image_sha256'],code+' source JPEG')
        original=cv2.imread(str(path),cv2.IMREAD_COLOR)
        need(original is not None and original.shape==(2160,3840,3),code+' original JPEG decode dimensions differ')
        cards=[]
        for index,seed,threshold,total,_ in SEEDS:
            row=by_seed[index][code]
            need(int(row['seed_index'])==index and float(row['frozen_operating_threshold'])==threshold, code+' seed metadata differs')
            boxes=row['boxes']
            need(len(boxes)==SIX_COUNTS[code][index-1],code+f' seed{index} predeclared box count differs')
            for seq,p in enumerate(boxes,1):
                need(p['within_frozen_ownership_region'] is True and p['passes_frozen_operating_threshold'] is True,
                     code+' saved operating prediction violates frozen filter')
                need(float(p['score'])>=threshold,code+' saved box below frozen threshold')
                card,m=make_card(original,p,code,index,seq)
                cards.append(card)
                all_meta.append(m)
        expected_total+=len(cards)
        # Exactly every preserved operating box, in frozen seed/order; no cherry-picking.
        for start in range(0,len(cards),PER_PAGE):
            subset=cards[start:start+PER_PAGE]
            rows_n=math.ceil(len(subset)/PAGE_COLS)
            atlas=np.zeros((rows_n*CARD_H,PAGE_COLS*CARD_W,3),dtype=np.uint8)
            for slot,card in enumerate(subset):
                ry=slot//PAGE_COLS; rx=slot%PAGE_COLS
                atlas[ry*CARD_H:(ry+1)*CARD_H,rx*CARD_W:(rx+1)*CARD_W,:]=card
            page_num=start//PER_PAGE+1
            name=f'{code}__native_prediction_crops_page{page_num:02d}.png'
            page_path=out_pages/name
            png_write(page_path,atlas)
            atlas_manifest.append({'sample_code':code,'page':page_num,'first_prediction_1_based':start+1,
                                   'last_prediction_1_based':start+len(subset),'prediction_crops':len(subset),
                                   'filename':name,'sha256':sha(page_path),'height':int(atlas.shape[0]),
                                   'width':int(atlas.shape[1]),'native_unscaled':True})
        print('SAMPLE',code,'boxes',len(cards),'atlas pages',math.ceil(len(cards)/PER_PAGE),flush=True)
        del original,cards

    need(len(all_meta)==expected_total==74,'Expected exactly 74 predeclared operating prediction crops')
    need(len(atlas_manifest)==sum(math.ceil(sum(SIX_COUNTS[c])/PER_PAGE) for c in SIX),'Atlas page completeness mismatch')
    counts_csv=OUT/'TIOMAN_NATIVE_PREDICTION_REVIEW_BLANK_WORKSHEET_v1.csv'
    fields=['sample_code','seed_index','box_index_in_image_seed','score','tile_id','image_box_xyxy','crop_xyxy',
            'annotation_overlap_visual_judgment','surface_texture_response_visual_judgment','distinct_structure_visual_judgment',
            'possible_artifact_notes','uncertainty_notes','reviewer','review_utc']
    with counts_csv.open('w',encoding='utf-8',newline='') as f:
        writer=csv.DictWriter(f,fieldnames=fields,lineterminator='\n'); writer.writeheader()
        for r in all_meta:
            writer.writerow({k:(json.dumps(r[k]) if k in ('image_box_xyxy','crop_xyxy') else r.get(k,'')) for k in fields})

    roi_manifest=OUT/'TIOMAN_NATIVE_PREDICTION_CROP_INDEX_v1.json'
    dumpj(roi_manifest,{'status':'ALL_74_PREDECLARED_OPERATING_BOXES_INCLUDED','index':all_meta,'pages':atlas_manifest,
                        'scientific_scope':'Unannotated Tioman visual inspection; no GT; no FTIR box identification',
                        'native_pixel_crop_shape':[384,384],'resized':False})
    overview=OUT/'README_NATIVE_RESOLUTION_REVIEW_v1.txt'
    overview.write_text(
        'DASO-Net Tioman native-resolution visual evidence\n\n'
        'Six samples were selected in the frozen protocol BEFORE inference.\n'
        'All 74 operating-threshold predictions from those six images are included, with no selection of individual boxes.\n'
        'Each atlas card shows a 384x384 native-pixel original crop on the left and the exact same crop with a rendered yellow prediction rectangle plus magenta center marker on the right.\n'
        'Cards are derived visual aids; rendered marks are NOT ground truth.\n'
        'The original source images, frozen inference, thresholds, source predictions, and 207 MB evidence ZIP are not modified by this audit.\n'
        'Measurement graphics and scale bars can be present in original source images; crop annotations do not classify their origin.\n'
        'The blank CSV is for qualified manual qualitative notes, not object-level true/false labels.\n'
        'No chemical polymer identity follows from image predictions or sample-linked FTIR library hits.\n'
        'Historical derived JSON hash mismatches and prior failed inference attempts remain documented elsewhere.\n',
        encoding='utf-8')

    # Portable *derived-review* ZIP: figures, native atlas pages, blank worksheet,
    # provenance; does NOT repack the frozen 207 MB results archive.
    source_items=sorted(p for p in OUT.rglob('*') if p.is_file() and p.name!='TIOMAN_NATIVE_RESOLUTION_REVIEW_EVIDENCE_v1.zip')
    index=[]
    for p in source_items:
        index.append({'path':p.relative_to(OUT).as_posix(),'sha256':sha(p),'bytes':p.stat().st_size})
    package_manifest=OUT/'TIOMAN_NATIVE_REVIEW_SOURCE_FILE_MANIFEST_v1.json'
    dumpj(package_manifest,{'files':index,'manifest_self_excluded':True})
    source_items.append(package_manifest)
    zip_path=OUT/'TIOMAN_NATIVE_RESOLUTION_REVIEW_EVIDENCE_v1.zip'
    temporary=OUT/'TIOMAN_NATIVE_RESOLUTION_REVIEW_EVIDENCE_v1.zip.incomplete'
    with zipfile.ZipFile(temporary,'w',compression=zipfile.ZIP_DEFLATED,compresslevel=6,allowZip64=True) as z:
        for p in source_items:
            z.write(p,arcname=p.relative_to(OUT).as_posix())
    with zipfile.ZipFile(temporary,'r') as z:
        need(z.testzip() is None,'ZIP CRC verification failed')
        need(set(z.namelist())=={p.relative_to(OUT).as_posix() for p in source_items},'Review ZIP inventory differs')
        for p in source_items:
            name=p.relative_to(OUT).as_posix()
            need(hashlib.sha256(z.read(name)).hexdigest()==sha(p),'Review ZIP readback SHA failure: '+name)
    os.replace(temporary,zip_path)
    package_sha=sha(zip_path)
    (OUT/'TIOMAN_NATIVE_RESOLUTION_REVIEW_EVIDENCE_v1.zip.sha256.txt').write_text(
        f'{package_sha}  {zip_path.name}\n',encoding='utf-8')

    record={
      'status':'PASS_ALL_74_PREDECLARED_BOXES_NATIVE_PIXEL_REVIEW_READY',
      'created_utc':datetime.now(timezone.utc).isoformat(),
      'source_run_state_sha256':PIN['run'][1],
      'independent_results_audit_sha256':PIN['audit'][1],
      'source_evidence_zip_sha256':PIN['portable'][1],
      'predeclared_visual_prep_sha256':PIN['visual_prep'][1],
      'six_codes':list(SIX),'box_count':len(all_meta),'atlas_pages':len(atlas_manifest),
      'review_csv_sha256':sha(counts_csv),'crop_index_sha256':sha(roi_manifest),
      'derived_zip_sha256':package_sha,
      'neither_existing_results_nor_protocol_modified':True,
      'model_loading_or_inference':False,'threshold_tuning':False,
      'ground_truth_or_ftir_identity_assigned':False,
      'historical_json_discrepancies_preserved':True,
    }
    status_path=OUT/'TIOMAN_NATIVE_RESOLUTION_REVIEW_PREPARATION_v1.json'
    dumpj(status_path,record)
    print('='*78)
    print('NATIVE REVIEW STATUS:',record['status'])
    print('NATIVE PREDICTION CROPS:',len(all_meta))
    print('CROP ATLAS PAGES:',len(atlas_manifest))
    print('BLANK MANUAL REVIEW ROWS:',len(all_meta))
    print('NATIVE REVIEW PACKAGE:',zip_path)
    print('NATIVE REVIEW ZIP SHA256:',package_sha)
    print('REVIEW STATUS JSON:',status_path)
    print('REVIEW STATUS SHA256:',sha(status_path))
    print('All frozen scientific artifacts unchanged. No inference, tuning, or GT evaluation.')
    print('='*78)
except BaseException as e:
    failure=OUT/'TIOMAN_NATIVE_REVIEW_FAILURE_v1.json'
    dumpj(failure,{'status':'FAILED_DERIVED_REVIEW_ATTEMPT_PRESERVED',
                   'error_type':type(e).__name__,'error':str(e),
                   'traceback':traceback.format_exc()[-12000:],
                   'source_inputs_read_only':True})
    print('STOP. PRESERVED DERIVED REVIEW FAILURE:',failure)
    raise
```

```python

# DASO-Net — FINAL EOD DOWNLOAD AND PRESERVATION
# Run in TODAY'S ORIGINAL Kaggle notebook only.
# No inference, training, calibration or tuning.

from pathlib import Path
from datetime import datetime, timezone
from IPython.display import FileLink, display
import hashlib
import json
import zipfile

ROOT = Path("/kaggle/working")

RESULTS_ZIP = (
    ROOT
    / "dasonet_tioman_audited_results_portable_freeze_v1"
    / "DASO_Net_Tioman_Audited_Results_Evidence_v1.zip"
)

NATIVE_ZIP = (
    ROOT
    / "dasonet_tioman_native_resolution_visual_audit_v1"
    / "TIOMAN_NATIVE_RESOLUTION_REVIEW_EVIDENCE_v1.zip"
)

EXPECTED = {
    RESULTS_ZIP:
        "2cad35c34469d5d59863b7f8456535003ba8062c426c7cdeecd8ffd5ca25cfc2",
    NATIVE_ZIP:
        "a0eba31d89542c59a757b03245a1638294d3f2a00dc9ffeaf95edeb2ec19ccde",
}

FAILURES = [
    (
        "initial",
        "dasonet_tioman_frozen_three_seed_inference_v1",
        "c90f241e722540ecbad566739daeaf7a2a636da3ebe651ea59c9783eaaaa144e",
    ),
    (
        "resume01",
        "dasonet_tioman_frozen_three_seed_inference_v1_resume01",
        "b833af6d52f5c5f07298256fc07e55e3d2c432010895fa13e99a994d133a06fa",
    ),
    (
        "resume02",
        "dasonet_tioman_frozen_three_seed_inference_v1_resume02",
        "e799985eae609614bcbba1150594a9ecdcaa57e6c3de1dae1bc948372d212e0d",
    ),
]

SMALL_EVIDENCE = [
    ROOT
    / "dasonet_tioman_native_resolution_visual_audit_v1"
    / "TIOMAN_NATIVE_RESOLUTION_REVIEW_PREPARATION_v1.json",

    ROOT
    / "dasonet_tioman_predeclared_visual_review_v1"
    / "TIOMAN_PREDECLARED_VISUAL_REVIEW_PREPARATION_v1.json",

    ROOT
    / "dasonet_tioman_results_integrity_audit_v1"
    / "TIOMAN_THREE_SEED_RESULTS_INTEGRITY_AUDIT_v1.json",

    ROOT
    / "dasonet_tioman_preprotocol_interface_audit_v1_resume02"
    / "TIOMAN_PREPROTOCOL_INTERFACE_AUDIT_RESUME02.json",

    ROOT
    / "dasonet_tioman_execution_interface_capture_v1"
    / "TIOMAN_FROZEN_EXECUTION_INTERFACE_CAPTURE_v1.json",

    ROOT
    / "dasonet_tioman_frozen_three_seed_inference_v1_resume03"
    / "TIOMAN_THREE_SEED_RUN_STATE_v1.json",
]


def ensure(condition, message):
    if not condition:
        raise RuntimeError(message)


def sha(path):
    h = hashlib.sha256()

    with Path(path).open("rb") as f:
        for piece in iter(
            lambda: f.read(4 * 1024 * 1024),
            b"",
        ):
            h.update(piece)

    return h.hexdigest()


def verify(path, digest, label):
    ensure(
        path.is_file(),
        "Missing " + label + ": " + str(path),
    )

    actual = sha(path)

    ensure(
        actual == digest,
        f"SHA mismatch in {label}: "
        f"{actual} != {digest}",
    )

    return actual


def new_output_dir():
    base = (
        ROOT
        / "dasonet_today_final_eod_download_verification_v1"
    )

    for i in range(100):
        candidate = (
            base if i == 0
            else Path(str(base) + f"_resume{i:02d}")
        )

        if not candidate.exists():
            candidate.mkdir(
                parents=True,
                exist_ok=False,
            )
            return candidate

    raise RuntimeError(
        "No unused EOD folder. Previous outputs preserved."
    )


print("=" * 76)
print("DASO-Net — FINAL ORIGINAL-NOTEBOOK EOD DOWNLOAD CHECK")
print("NO inference, training, model loading or tuning")
print("=" * 76)

# ============================================================
# A. Verify the two essential evidence ZIPs.
# ============================================================

for path, expected_sha in EXPECTED.items():
    verify(path, expected_sha, path.name)

    sidecar = path.with_name(
        path.name + ".sha256.txt"
    )

    ensure(
        sidecar.is_file(),
        "Missing SHA sidecar: " + str(sidecar),
    )

    sidecar_sha = (
        sidecar.read_text(
            encoding="utf-8"
        ).split()[0]
    )

    ensure(
        sidecar_sha == expected_sha,
        "SHA sidecar mismatch: " + str(sidecar),
    )

    with zipfile.ZipFile(path) as archive:
        ensure(
            archive.testzip() is None,
            "ZIP integrity failure: " + str(path),
        )

        print(
            "PASS",
            path.name,
            "| members =", len(archive.namelist()),
            "| bytes =", path.stat().st_size,
        )

# ============================================================
# B. Preserve earlier failed attempts.
# ============================================================

sources = {}

for name, folder, expected_sha in FAILURES:
    base = ROOT / folder

    failure = (
        base / "TIOMAN_THREE_SEED_FAILURE_v1.json"
    )

    state = (
        base / "TIOMAN_THREE_SEED_RUN_STATE_v1.json"
    )

    verify(
        failure,
        expected_sha,
        f"{name} failure report",
    )

    ensure(
        state.is_file(),
        f"Missing preserved {name} state: {state}",
    )

    sources[f"failures/{name}/failure.json"] = failure
    sources[f"failures/{name}/run_state.json"] = state

for path in SMALL_EVIDENCE:
    ensure(
        path.is_file(),
        "Missing provenance file: " + str(path),
    )

    relative = path.relative_to(ROOT).as_posix()

    sources[
        "session_provenance/" + relative
    ] = path

# ============================================================
# C. Package compact failure/provenance evidence.
# ============================================================

out = new_output_dir()

trace_zip = (
    out
    / "DASONET_TODAY_FAILED_RUNS_AND_AUDIT_TRAIL_v1.zip"
)

manifest = {
    "purpose": (
        "Preserve the three failed inference attempts "
        "and small provenance records."
    ),
    "created_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "source_inputs_modified": False,
    "model_inference_or_training": False,
    "successful_results_zip_sha256": (
        EXPECTED[RESULTS_ZIP]
    ),
    "native_review_zip_sha256": (
        EXPECTED[NATIVE_ZIP]
    ),
    "files": [
        {
            "zip_path": name,
            "size_bytes": path.stat().st_size,
            "sha256": sha(path),
        }
        for name, path in sorted(sources.items())
    ],
}

manifest_bytes = (
    json.dumps(
        manifest,
        sort_keys=True,
        indent=2,
    ) + "\n"
).encode("utf-8")

with zipfile.ZipFile(
    trace_zip,
    "w",
    compression=zipfile.ZIP_DEFLATED,
    compresslevel=6,
) as archive:

    for name, path in sorted(sources.items()):
        archive.write(
            path,
            arcname=name,
        )

    archive.writestr(
        "MANIFEST_SHA256_v1.json",
        manifest_bytes,
    )

# ============================================================
# D. Independently verify the supplementary archive.
# ============================================================

with zipfile.ZipFile(trace_zip, "r") as archive:
    ensure(
        archive.testzip() is None,
        "Supplementary ZIP CRC failure.",
    )

    ensure(
        len(archive.namelist()) == len(sources) + 1,
        "Supplementary ZIP file count mismatch.",
    )

    ensure(
        archive.read("MANIFEST_SHA256_v1.json")
        == manifest_bytes,
        "Supplementary ZIP manifest mismatch.",
    )

    for entry in manifest["files"]:
        actual = hashlib.sha256(
            archive.read(entry["zip_path"])
        ).hexdigest()

        ensure(
            actual == entry["sha256"],
            "Archived member SHA mismatch: "
            + entry["zip_path"],
        )

trace_sha = sha(trace_zip)

trace_sidecar = trace_zip.with_name(
    trace_zip.name + ".sha256.txt"
)

trace_sidecar.write_text(
    f"{trace_sha}  {trace_zip.name}\n",
    encoding="utf-8",
)

# ============================================================
# E. Display download links.
# ============================================================

print()
print("=" * 76)
print(
    "EOD PRESERVATION STATUS:",
    "PASS_THREE_ARCHIVES_READY_TO_DOWNLOAD",
)

print("RESULTS ZIP SHA256:", EXPECTED[RESULTS_ZIP])
print("NATIVE ZIP SHA256:", EXPECTED[NATIVE_ZIP])
print("FAILED-RUN TRAIL ZIP SHA256:", trace_sha)

print()
print("DOWNLOAD THESE THREE ZIPs AND THEIR SHA SIDECARS:")

for archive in (
    RESULTS_ZIP,
    NATIVE_ZIP,
    trace_zip,
):
    sidecar = archive.with_name(
        archive.name + ".sha256.txt"
    )

    print()
    print("ZIP:", archive)
    print("SHA:", sidecar)

    display(FileLink(str(archive)))
    display(FileLink(str(sidecar)))

print()
print("ALSO DOWNLOAD THE FINAL EOD HANDOVER FROM CHATGPT.")
print("DO NOT RUN THE NEW-NOTEBOOK RESTART CELL TODAY.")
print("All frozen scientific inputs remain unchanged.")
print("=" * 76)
```
